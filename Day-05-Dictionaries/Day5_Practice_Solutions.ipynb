### A2. What is `d` at the end?

````python
d = {"a": 1}d["b"] = 2d["a"] = 99print(d)print(len(d))
````

*Your prediction:*  

> ### ✅ Answer A2> ```> {'a': 99, 'b': 2}> 2> ```> **Why:** `d["b"] = 2` adds a new key. `d["a"] = 99` finds an existing key and overwrites> it — the original `1` is simply gone, with no warning.>> The length is 2, not 3: keys are unique, so reassigning `"a"` did not add anything.>> Note the **order**: `'a'` still comes first because it was inserted first. Dictionaries> keep insertion order since Python 3.7.

### A3. True or False for each?

````python
marks = {"Python": 88, "Stats": 71}print("Python" in marks)print(88 in marks)print(88 in marks.values())
````

*Your prediction:*  

> ### ✅ Answer A3> ```> True> False> True> ```> **Why:** `in` on a dictionary checks the **keys** only — never the values.> `88` is a value, not a key, so the second line is `False`.>> To search values you must say so explicitly: `88 in marks.values()`.>> This catches almost everyone the first time. It is worth a show of hands before revealing.

### A5. What is `counts` at the end?

````python
words = ["a", "b", "a"]counts = {}for w in words:    counts[w] = counts.get(w, 0) + 1print(counts)
````

*Your prediction:*

### A6. What does this produce?

### A7. Which of these are legal?

````python
a = {"name": 1}b = {42: 1}c = {(3, 7): 1}d = {[1, 2]: 1}
````

*Your prediction:*  

> ### ✅ Answer A7> The first three are fine. The fourth raises:> ```> TypeError: unhashable type: 'list'> ```> **Why:** keys must be **immutable**. A dictionary locates a value by computing a number> from the key; if the key could change afterwards, that number would no longer match and> the value would be lost inside the dictionary.>> So lists, sets and dictionaries are refused as keys. Strings, numbers and tuples are fine.>> **This is the practical reason tuples exist** — the answer to yesterday's *"why would I> ever use a tuple?"*

### A8. What does the merge give?

````python
defaults = {"colour": "blue", "size": 10}user     = {"size": 14}print(defaults | user)print(user | defaults)
````

*Your prediction:*  

> ### ✅ Answer A8> ```> {'colour': 'blue', 'size': 14}> {'size': 10, 'colour': 'blue'}> ```> **Why:** on a clashing key, the **right-hand side wins**. In the first line the user's> `size` of 14 overrides the default of 10 — which is what you want.>> The second line is the bug: putting defaults on the right throws the user's choice away.>> **Rule:** when merging settings, the more specific dictionary goes second.

### B2. This should count the words. It doesn't.

````python
words = ["a", "b", "a"]counts = {}for w in words:    counts[w] = counts[w] + 1print(counts)
````

*Error:* *Your fix:*

> ### ✅ Answer B2> **Error:** `KeyError: 'a'` — on the very first word.> **Cause:** the first time a word appears there is no entry yet, so `counts[w]` has> nothing to read. You cannot add one to something that does not exist.

````python
words = ["a", "b", "a"]counts = {}for w in words:    counts[w] = counts.get(w, 0) + 1      # 0 is the starting valueprint(counts)          # {'a': 2, 'b': 1}# Also valid, and worth seeing:from collections import defaultdictcounts = defaultdict(int)                  # missing keys start at 0for w in words:    counts[w] += 1print(dict(counts))
````

### B3. This should delete the high scores.

````python
marks = {"a": 30, "b": 80, "c": 90}for k in marks:    if marks[k] > 50:        del marks[k]print(marks)
````

*Error:* *Your fix:*

> ### ✅ Answer B3> **Error:** `RuntimeError: dictionary changed size during iteration`> **Cause:** you cannot add or remove entries while looping over the dictionary itself.> The loop is walking the dictionary, and deleting pulls the ground out from under it.

````python
marks = {"a": 30, "b": 80, "c": 90}# FIX 1 - loop over a fixed snapshot of the keysfor k in list(marks):          # list() copies the keys first    if marks[k] > 50:        del marks[k]print(marks)                   # {'a': 30}# FIX 2 - build a new dictionary instead of editing in place (often cleaner)marks = {"a": 30, "b": 80, "c": 90}kept = {}for k, v in marks.items():    if v <= 50:        kept[k] = vprint(kept)                    # {'a': 30}
````
