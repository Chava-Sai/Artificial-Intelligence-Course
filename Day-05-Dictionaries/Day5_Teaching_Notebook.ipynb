---# 1. The basics## 1.1 A dictionary maps a key to a value```    KEY          VALUE  "name"   ──►   "Ravi"  "age"    ──►    23  "city"   ──►   "Pune"```Keys are **unique**. Values can be anything, and can repeat.

````python
# A list stores by POSITIONstudent_list = ["Ravi", 23, "Pune"]print(student_list[1])        # 23 - but 23 of what?# A dictionary stores by NAMEstudent = {"name": "Ravi", "age": 23, "city": "Pune"}print(student["age"])         # 23 - unmistakably the ageprint(type(student), len(student))
````

The real difference is not syntax — it is that **the data explains itself**.With a list you must remember that position 1 holds the age. With a dictionary the data says so.

## 1.3 Reading — and the `KeyError`

````python
marks = {"Python": 88, "Stats": 71}print(marks["Python"])        # 88try:    print(marks["ML"])except KeyError as e:    print("KeyError:", e)     # it STOPS your program - not a warning
````

````python
# .get() is the safe way to readprint(marks.get("ML"))          # None  - no crashprint(marks.get("ML", 0))       # 0     - your own defaultprint(marks.get("Python", 0))   # 88print("ML" in marks)            # False - check before reading
````

> ### Which should you use?>> Use **square brackets** when the key *must* be there and its absence is a genuine bug> you want to hear about immediately.>> Use **`.get()` with a default** when the key is genuinely optional — a config setting,> or a field a web API may not have sent.>> Choosing between them deliberately is what separates code that fails loudly from code> that fails at 2am.

````python
marks = {"Python": 88}marks["Stats"] = 71       # key is new     -> ADDSmarks["Python"] = 92      # key exists     -> UPDATESprint(marks)# One syntax does both. There is no warning when you overwrite.
````

````python
# To add ONLY if the key is missing, use setdefaultd = {"a": 1}d.setdefault("a", 99)     # "a" exists - left aloned.setdefault("b", 99)     # "b" is new  - insertedprint(d)                  # {'a': 1, 'b': 99}
````

````python
d = {"a": 1, "b": 2, "c": 3}del d["a"]                    ; print("del        ->", d)value = d.pop("b")            ; print("pop        ->", d, "| returned", value)missing = d.pop("zzz", None)  ; print("pop default->", missing, "(no error)")try:    d.pop("zzz")except KeyError as e:    print("pop without default -> KeyError:", e)
````

---# 2. Iterating## 2.1 keys, values and items

````python
marks = {"Python": 88, "Stats": 71, "ML": 64}print("Looping directly gives the KEYS:")for subject in marks:    print(" ", subject)print("\n.values() gives the values:")for m in marks.values():    print(" ", m)print("\n.items() gives BOTH:")for subject, m in marks.items():    print(f"  {subject:8} {m}")
````

`.items()` is the one you will use most. Each item is a `(key, value)` tuple and the loopunpacks it — Day 4's tuple unpacking doing real work. It is the same shape as `enumerate()`and `zip()` from Day 2.

````python
# A realistic usemarks = {"Python": 88, "Stats": 71, "ML": 34}for subject, m in marks.items():    status = "pass" if m >= 40 else "fail"    print(f"{subject:8} {m:3}  {status}")print("\nTotal:", sum(marks.values()))print("Best  :", max(marks.values()))print("Count :", len(marks))
````

## 2.2 ⚠️ `in` checks KEYS, never values

````python
marks = {"Python": 88, "Stats": 71}print("'Python' in marks      ->", "Python" in marks)      # Trueprint("88 in marks            ->", 88 in marks)            # False!print("88 in marks.values()   ->", 88 in marks.values())   # True
````

## 2.3 Dictionaries keep insertion orderSince Python 3.7, looping a dictionary gives keys back in the order you added them.Older tutorials say dictionaries are unordered — that advice is out of date.It is also why `dict.fromkeys()` could deduplicate *while keeping order* yesterday.

````python
d = {}d["z"] = 1d["a"] = 2d["m"] = 3print(list(d))        # ['z', 'a', 'm'] - insertion order, not sorted# To get them sorted, ask for it explicitlyprint(sorted(d))                    # sorted keysprint(sorted(d.items()))            # sorted by key
````

### ⚠️ Never add or delete while looping

````python
d = {"a": 1, "b": 2, "c": 3}try:    for k in d:        if d[k] > 1:            del d[k]except RuntimeError as e:    print("RuntimeError:", e)
````

````python
# THE FIX: loop over a snapshot of the keysd = {"a": 1, "b": 2, "c": 3}for k in list(d):          # list() takes a fixed copy of the keys    if d[k] > 1:        del d[k]print(d)                   # {'a': 1}
````

Methods and patterns| Method | What it does ||---|---|| `get(k, default)` | read safely, never raises || `keys()` / `values()` / `items()` | views of keys / values / pairs || `setdefault(k, v)` | read `k`, inserting `v` first if absent || `update(other)` | merge another dict in, overwriting on clashes || `pop(k, default)` | remove and return the value || `copy()` | a shallow copy — Day 4's lesson applies |> `keys()`, `values()` and `items()` return **views**, not lists. A view stays connected to> the dictionary — add an entry and the view shows it. Wrap it in `list()` for a fixed snapshot.

````python
d = {"a": 1}view = d.keys()print("view before:", view)d["b"] = 2print("view after :", view)     # the view updated itselfsnapshot = list(d.keys())d["c"] = 3print("snapshot   :", snapshot) # frozen at the moment it was taken
````

## 3.1 The counting pattern**How many times does each item appear?** This is the single most reused dictionary idiom.

````python
words = ["a", "b", "a", "c", "a"]# The obvious attempt FAILScounts = {}try:    for w in words:        counts[w] = counts[w] + 1except KeyError as e:    print("KeyError:", e, "- there is nothing to add to the first time we see a word")
````

````python
# THE PATTERN: .get(w, 0) supplies a starting value of zerowords = ["a", "b", "a", "c", "a"]counts = {}for w in words:    counts[w] = counts.get(w, 0) + 1print(counts)      # {'a': 3, 'b': 1, 'c': 1}
````

````python
# A real use: letter frequencysentence = "data science is fun"freq = {}for ch in sentence:    if ch != " ":        freq[ch] = freq.get(ch, 0) + 1# sorted by count, highest first - Day 4's sorted() with a keyfor ch, n in sorted(freq.items(), key=lambda pair: pair[1], reverse=True)[:5]:    print(f"{ch!r}: {n}")
````

### Python has a built-in — but write the loop first

````python
from collections import Counterwords = ["a", "b", "a", "c", "a"]print(Counter(words))print(Counter(words).most_common(2))
````

`Counter` does the same job in one line, and you should use it in real work. But write themanual loop a dozen times first — the pattern of *"look up with a default, then update"*appears everywhere, and `Counter` only solves this one instance of it.

## 3.2 The grouping pattern**Collect items that share a key.**

````python
records = [("maths", "Ravi"), ("cs", "Sara"), ("maths", "Amit")]groups = {}for subject, name in records:    groups.setdefault(subject, []).append(name)print(groups)      # {'maths': ['Ravi', 'Amit'], 'cs': ['Sara']}
````

### Reading that middle line`groups.setdefault(subject, [])` returns the list for that subject. If the subject is new,it inserts an empty list **first**, then returns it. Either way you get a list back, so`.append()` always works.One line replaces this:

````python
# The long form - what setdefault saves yourecords = [("maths", "Ravi"), ("cs", "Sara"), ("maths", "Amit")]groups = {}for subject, name in records:    if subject not in groups:        groups[subject] = []    groups[subject].append(name)print(groups)      # identical result
````

## 3.3 Merging dictionaries

````python
a = {"x": 1, "y": 2}b = {"y": 99, "z": 3}merged = {**a, **b}       # a new dictprint("unpacking:", merged)merged = a | b            # Python 3.9+, cleanestprint("| operator:", merged)a.update(b)               # changes a IN PLACEprint("update   :", a)
````

> ### ⚠️ On a clash, the right-hand side wins>> `{"a": 1} | {"a": 2}` gives `{"a": 2}`.>> Merging default settings with user settings? Put the **user's** dictionary second,> so their choices override your defaults.

````python
defaults = {"colour": "blue", "size": 10, "debug": False}user     = {"size": 14}settings = defaults | user      # user wins where they specified somethingprint(settings)
````

---# 4. Nested dictionariesThis is the shape real data actually arrives in.

````python
students = {    "ravi": {"age": 23, "marks": [88, 71]},    "sara": {"age": 22, "marks": [91, 84]},}print(students["ravi"]["age"])          # 23print(students["ravi"]["marks"][0])     # 88# Read left to right: one step per bracket.
````

````python
# A list of dictionaries - what a CSV becomes, and what an API returnsrows = [    {"name": "Ravi", "mark": 88},    {"name": "Sara", "mark": 91},    {"name": "Amit", "mark": 64},]for r in rows:    print(f"{r['name']:6} {r['mark']}")print("average:", sum(r["mark"] for r in rows) / len(rows))
````

### ⚠️ Deep access fails deep

````python
students = {"ravi": {"age": 23}}try:    print(students["neha"]["age"])except KeyError as e:    print("KeyError:", e)# THE FIX: chain .get() with an empty dict as the defaultprint(students.get("neha", {}).get("age"))          # None - no crashprint(students.get("neha", {}).get("age", "n/a"))   # n/aprint(students.get("ravi", {}).get("age"))          # 23
````

The empty `{}` default is the trick: if `"neha"` is missing you get an empty dictionaryback, and calling `.get()` on that returns `None` instead of crashing.

## 4.1 What can be a key?Keys must be **immutable** — because a dictionary finds a value by computing a number fromthe key. If the key could change afterwards, that number would no longer match and thevalue would be lost inside the dictionary.

````python
d = {"name": 1}       # str    okd = {42: 1}           # int    okd = {(3, 7): 1}       # tuple  okd = {True: 1}         # bool   okprint("all fine so far")try:    d = {[1, 2]: 1}except TypeError as e:    print("TypeError:", e)try:    d = {(1, [2]): 1}     # a tuple containing a list is also refusedexcept TypeError as e:    print("TypeError:", e)
````

This is the practical reason tuples exist. If you asked yesterday *"why would I ever usea tuple?"* — this is the answer.**Values have no such rule.** A value can be anything at all, including another dictionary,which is exactly what makes nested structures possible.

````python
d = {    "marks": [88, 71, 64],        # a list    "meta":  {"term": 2},         # a dict    "tags":  {"cs", "maths"},     # a set}print(d)
````

---# 5. Putting it together — a gradebook from raw records

````python
records = [    ("ravi", "python", 88), ("sara", "python", 91),    ("ravi", "stats", 71),  ("amit", "python", 64),    ("sara", "stats", 84),  ("ravi", "ml", 90),]def build_gradebook(rows):    """Group (student, subject, mark) rows into a nested dictionary."""    book = {}    for name, subject, mark in rows:            # unpack the tuple        book.setdefault(name, {})[subject] = mark    return bookbook = build_gradebook(records)print(book)print()for name, subjects in book.items():    marks = subjects.values()    avg   = sum(marks) / len(marks)    print(f"{name:6} {len(marks)} subjects   avg {avg:.1f}")print()print("Ravi's stats mark:", book["ravi"].get("stats"))print("Amit's ml mark   :", book["amit"].get("ml", "not taken"))
````

What each piece contributes:- **Nested dict** — student → subject → mark- **`setdefault`** — creates the inner dictionary on demand- **Tuple unpacking** — three names from one row- **`.items()`** — loop over names and their marks together- **`.values()`** — feeds straight into `sum()`- **`.get()` with a default** — "not taken" instead of a crashSix flat rows become a structure you can query by name — the same reshaping pandas doesfor you later.
