### A5. Precedence check — no calculator.

````python
print(2 + 3 * 4 ** 2)print(2 ** 3 ** 2)print((2 + 3) * 4)print(10 - 4 - 3)
````

*Your prediction:*

### A6. Identity vs equality.

````python
a = [1, 2, 3]b = [1, 2, 3]c = aprint(a == b)print(a is b)print(a is c)
````

*Your prediction:*  

> ### ✅ Answer A6> ```> True> False> True> ```> **Why:** `==` compares **contents**, `is` compares **memory identity**.> `a` and `b` hold equal values but are two separate list objects.> `c = a` did not copy anything — it created a second name for the *same* object.>> Follow-up to ask the class: *"if I now run `c.append(4)`, what does `a` print?"* → `[1, 2, 3, 4]`.

### A7. Short-circuit evaluation. Does `check()` run?

````python
def check():    print("check() ran")    return Trueprint(False and check())print(True or check())print(True and check())
````

*Your prediction:*  

> ### ✅ Answer A7> ```> False> True> check() ran> True> ```> **Why:**> - `False and ...` → `and` already knows the answer is `False`, so the right side is **never evaluated**.> - `True or ...` → `or` already knows the answer is `True`, so the right side is **never evaluated**.> - `True and check()` → the left side is inconclusive, so `check()` **must** run. It prints, then returns `True`.>> This is why `if n != 0 and total/n > 5:` is safe — the division never happens when `n` is 0.

### A8. String vs number operators.

````python
print("5" + "3")print(5 + 3)print("ab" * 3)print(int("7") + int("3"))
````

*Your prediction:*  

> ### ✅ Answer A8> ```> 53> 8> ababab> 10> ```> **Why:** `+` on two strings **concatenates**; on two numbers it **adds**.> `*` on a string and an int **repeats** the string.> Casting with `int()` first gives real arithmetic.>> This is exactly the bug students hit with `input()` — it returns a string, so `input() + input()` glues text together instead of adding.
