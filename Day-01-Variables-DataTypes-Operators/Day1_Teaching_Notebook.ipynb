---# 1. Variables## 1.1 What actually happens on assignmentIn many languages a variable is a **box** that holds a value. In Python it is more accurate to saya variable is a **name tag (label) attached to an object in memory**.```    x = 10    x  ──────────►  ┌──────────┐   (name)           │    10    │   int object in memory                    └──────────┘```Three things happen when you write `x = 10`:1. Python creates the object `10` in memory.2. Python creates the name `x`.3. Python **binds** the name `x` to that object.The `=` sign is **not** "equals". It is **"assign"**. Read `x = 10` as *"x gets 10"*.

````python
x = 10print(x)print(type(x))   # what kind of object is it?print(id(x))     # the memory address the name points to
````

### Rebinding: variables are not permanentAssigning again does **not** change the object `10`. It points the name at a *different* object.

````python
x = 10print("x =", x, "| id:", id(x))x = 99print("x =", x, "| id:", id(x))   # different id -> a different object entirelyx = "now I am text"print("x =", x, "| type:", type(x))   # the SAME name can hold a different type
````

> ### Key idea: Python is **dynamically typed**> You never declare a type. The *value* carries the type, not the name.> In C or Java you would write `int x = 10;`. In Python, just `x = 10`.>> This is convenient, but it also means a typo can silently change your data's type.> That is a very common bug in data science code — we will see it in Section 3.

## 3.2 Implicit conversion (Python does it for you)When you mix `int` and `float`, Python quietly upgrades the `int` to a `float`so no precision is lost. This is called **type coercion**.

````python
result = 5 + 2.0      # int + floatprint(result, type(result))     # 7.0 <class 'float'>print(True + 4.5)     # bool -> int -> float
````

## 3.3 Explicit conversion (casting) — you do it yourself```        str "42"  ──int()──►  int 42  ──float()──►  float 42.0             ▲                   │                      │             └───────str()───────┴──────str()───────────┘```

````python
print(int("42"),       type(int("42")))print(float("3.75"),   type(float("3.75")))print(str(99),         type(str(99)))print(bool("data"),    bool(""))
````

### ⚠️ Three casting traps

````python
# TRAP 1: int() TRUNCATES a float, it does not roundprint(int(9.99))      # 9   (not 10!)print(int(-9.99))     # -9  (toward zero)print(round(9.99))    # 10  -- use round() when you want rounding
````

````python
# TRAP 2: int() cannot parse a decimal STRING directlytry:    print(int("3.5"))except ValueError as e:    print("ValueError:", e)print("Fix -> go through float first:", int(float("3.5")))
````

````python
# TRAP 3: garbage strings raise ValueError, not a silent NaNfor s in ["100", "  100  ", "1,00,000", "abc", ""]:    try:        print(f"{repr(s):>12} -> {int(s)}")    except ValueError as e:        print(f"{repr(s):>12} -> ValueError")
````

---# 4. OperatorsAn **operator** performs an operation on one or more **operands**.```        7      +      3        │      │      │     operand  op   operand```We cover five families today.

## 4.1 Arithmetic operators| Operator | Name | `a=17, b=5` | Result ||---|---|---|---|| `+` | Addition | `a + b` | `22` || `-` | Subtraction | `a - b` | `12` || `*` | Multiplication | `a * b` | `85` || `/` | **True division** — always returns `float` | `a / b` | `3.4` || `//` | **Floor division** — divides and rounds DOWN | `a // b` | `3` || `%` | **Modulus** — the remainder | `a % b` | `2` || `**` | Exponent (power) | `a ** 2` | `289` |

````python
a, b = 17, 5print("a + b  =", a + b)print("a - b  =", a - b)print("a * b  =", a * b)print("a / b  =", a / b,  type(a / b))    # ALWAYS float, even for 10/5print("a // b =", a // b)print("a %  b =", a % b)print("a ** 2 =", a ** 2)
````

````python
# Note: / always gives a float, even when it divides evenlyprint(10 / 5, type(10 / 5))     # 2.0  <class 'float'>print(10 // 5, type(10 // 5))   # 2    <class 'int'>
````

### `//` and `%` are more useful than they look`//` and `%` together split a number into "how many whole groups" and "what is left over".

````python
total_minutes = 197hours = total_minutes // 60     # how many whole hoursmins  = total_minutes % 60      # leftover minutesprint(f"{total_minutes} minutes = {hours} hr {mins} min")
````

````python
# The classic use of % : test divisibility (even / odd)for n in [10, 7, 24, 33]:    print(n, "is", "EVEN" if n % 2 == 0 else "ODD")
````

````python
# CAREFUL: floor division rounds DOWN (toward negative infinity), not toward zeroprint( 7 // 2)     #  3print(-7 // 2)     # -4   (not -3!)print( 7 %  3)     #  1print(-7 %  3)     #  2   (Python's % follows the sign of the divisor)
````

## 4.2 Operator precedence — who goes first?Python evaluates in this order (top = highest priority):```  1.  ()                    parentheses  2.  **                    exponent            (right-to-left!)  3.  +x  -x  ~x            unary plus/minus  4.  *  /  //  %           multiply / divide  5.  +  -                  add / subtract  6.  <  <=  >  >=  ==  !=  comparisons  7.  not  8.  and  9.  or```**Practical advice: when in doubt, use parentheses.** Clear beats clever.

````python
print(2 + 3 * 4)        # 14, not 20  -> * before +print((2 + 3) * 4)      # 20          -> parentheses winprint(2 ** 3 ** 2)      # 512, not 64 -> ** is RIGHT-to-left: 2**(3**2) = 2**9print((2 ** 3) ** 2)    # 64print(10 - 4 - 3)       # 3  -> left-to-right for same precedence
````

## 4.7 Identity operators — `is`, `is not`- `==` asks: **do these have the same VALUE?**- `is` asks: **are these the exact SAME OBJECT in memory?**Use `is` only for `None`, `True`, `False`. Use `==` for everything else.

````python
a = [1, 2, 3]b = [1, 2, 3]      # a separate list with equal contentsc = a              # the SAME list, second nameprint("a == b :", a == b)    # True  -> same valuesprint("a is b :", a is b)    # False -> different objectsprint("a is c :", a is c)    # True  -> same objectprint("id(a):", id(a))print("id(b):", id(b))print("id(c):", id(c))
````

````python
# Mutating through one name changes what the other name sees -- because it is ONE objectc.append(99)print("a is now:", a)     # a changed too!
````

### ⚠️ Why `is` on numbers is a trapPython caches small integers (-5 to 256), so `is` sometimes *looks* like it works.It is an implementation detail — never rely on it.

````python
x = 256; y = 256print("256 is 256 ->", x is y)      # True (cached)x = 257; y = 257print("257 is 257 ->", x is y)      # often False -- same value, different objectsprint("257 == 257 ->", 257 == 257)  # ALWAYS True. Use == for values.
````
