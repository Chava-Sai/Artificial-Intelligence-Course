### A1. What is printed?

````python
s = pd.Series([88, 71, 64], index=["Ravi", "Sara", "Amit"])
print(s["Sara"])
print(s.index.tolist())
````

*Your prediction:*  

> ### ✅ Answer A1
> ```
> 71
> ['Ravi', 'Sara', 'Amit']
> ```
> **Why:** with a custom index, you look values up by **label**, exactly like Day 5's
> dictionary. The index is the list of labels.
>
> A Series behaves like a dict *and* like a NumPy array at the same time — that combination
> is the reason pandas exists.

### A2. What does each line give?

````python
s = pd.Series([88, 71, 64], index=["Ravi", "Sara", "Amit"])
print(s * 2)
print(s[s > 70])
````

*Your prediction:*  

> ### ✅ Answer A2
> ```
> Ravi    176
> Sara    142
> Amit    128
> dtype: int64
> Ravi    88
> Sara    71
> dtype: int64
> ```
> **Why:** `s * 2` is Day 13's broadcasting — the 2 is stretched across every value.
> `s[s > 70]` is Day 12's boolean mask. **The labels travel with the values** through both
> operations, which is something a bare NumPy array could never do.

### A3. Same four numbers — what does each library give?

````python
arr = np.array([88., 71., np.nan, 64.])
print(arr.mean())
print(pd.Series(arr).mean())
````

*Your prediction:*  

> ### ✅ Answer A3
> ```
> nan
> 74.33333333333333
> ```
> **This reverses yesterday's lesson.** NumPy returns `nan` because one value is missing.
> pandas **skips the missing value by default** and averages the other three.
>
> pandas was built for real tables, where gaps are normal — returning `nan` for every column
> with one blank would make it nearly useless. But the convenience has a cost: pandas stays
> **silent**, so a mean over 3 of 4 values looks exactly like a mean over all 4.
>
> `pd.Series(arr).mean(skipna=False)` gives NumPy's behaviour back.

### A4. Are these the same?

````python
arr = np.array([88., 71., 64.])
print(round(arr.std(), 4))
print(round(pd.Series(arr).std(), 4))
````

*Your prediction:*  

> ### ✅ Answer A4
> ```
> 10.0775
> 12.3423
> ```
> **They differ.** NumPy's `std` defaults to `ddof=0` (population, divides by *n*). pandas
> defaults to `ddof=1` (sample, divides by *n − 1*).
>
> This is exactly how two people get different numbers from the same data and each assume
> the other made a mistake. **State `ddof` explicitly** whenever the answer matters.

### A5. What type comes back?

````python
print(type(df["python"]).__name__)
print(type(df[["python"]]).__name__)
print(type(df[["name", "python"]]).__name__)
````

*Your prediction:*  

> ### ✅ Answer A5
> ```
> Series
> DataFrame
> DataFrame
> ```
> **Why:** single brackets with one name give a **Series**. Double brackets — a Python list
> of names inside the selector — always give a **DataFrame**, even when the list holds only
> one name.
>
> This matters because they behave differently: a Series has one dtype and one name, a
> DataFrame has columns. Many later functions want one specifically.

### A6. Does this work?

*Your prediction:*

### A7. What are the shape and dtypes?

### B2. This should total the `count` column.

````python
inventory = pd.DataFrame({"item": ["pen", "book"], "count": [12, 5]})
print(inventory.count.sum())
````

*What's wrong:* 

*Your fix:*

> ### ✅ Answer B2
> **Symptom:** `AttributeError: 'function' object has no attribute 'sum'`
> **Cause:** `inventory.count` is not your column — it is the DataFrame's built-in `count()`
> **method**, which counts non-missing values. Attribute access finds the method first.
>
> The same trap exists for any column called `sum`, `mean`, `max`, `min`, `size`, `shape`
> and many others.

````python
inventory = pd.DataFrame({"item": ["pen", "book"], "count": [12, 5]})

print(inventory["count"].sum())       # 17 - bracket access always finds the column

print(type(inventory.count).__name__)     # method
print(type(inventory["count"]).__name__)  # Series

# The rule: always use df["col"]. It works for every name, including ones with
# spaces, which attribute access cannot handle at all.
````

### B3. This works — but it should not be written this way.

````python
totals = []
for i in range(len(df)):
    totals.append(df["python"][i] + df["stats"][i])
df["total"] = totals
print(df[["name", "total"]])
````

*What's wrong:* 

*Your fix:*

> ### ✅ Answer B3
> **Symptom:** correct output, so nothing seems wrong.
> **Cause:** it loops over the rows in Python. On five rows you will not notice; on a real
> dataset of a million rows it is hundreds of times slower than the one-line version, and
> the code is four times as long.

````python
df["total"] = df["python"] + df["stats"]      # one line, whole columns
print(df[["name", "total"]])

# Exactly Day 13's rule: if you reach for  for i in range(len(...)),
# stop and ask what the whole-column expression would be.
````

### D1. A messy CSV
Load the file below. Then answer: **what is the true average salary?** Be careful — inspect
before you calculate.

````python
with open("staff.csv", "w") as f:
    f.write("name,dept,salary\n")
    f.write("Ravi,Sales,52000\n")
    f.write("Sara,sales,61000\n")
    f.write("Amit,IT,\n")
    f.write("Neha,IT,58000\n")
    f.write("Kiran,Sales ,49000\n")

# your code here
````

````python
with open("staff.csv", "w") as f:
    f.write("name,dept,salary\n")
    f.write("Ravi,Sales,52000\n")
    f.write("Sara,sales,61000\n")
    f.write("Amit,IT,\n")
    f.write("Neha,IT,58000\n")
    f.write("Kiran,Sales ,49000\n")

staff = pd.read_csv("staff.csv")

# ---- 1. INSPECT first
staff.info()
print()
print(staff["dept"].value_counts())

# THREE problems, all invisible until you look:
#   - salary has 1 missing value (Amit), and became float64
#   - "Sales", "sales" and "Sales " are three DIFFERENT values
#   - a naive mean would silently skip Amit

print()
print("naive mean     :", staff["salary"].mean())
print("computed from  :", staff["salary"].count(), "of", len(staff), "staff")

# ---- 2. Report it honestly
print(f"\nAverage salary is {staff['salary'].mean():,.0f}, "
      f"from {staff['salary'].count()} of {len(staff)} staff "
      f"({staff['salary'].isna().sum()} salary missing).")

# ---- 3. The department labels need cleaning before any per-dept analysis
#
# A TRAP worth seeing: the obvious fix breaks something new.
print()
print("with .str.title():", staff["dept"].str.strip().str.title().value_counts().to_dict())
#   -> {'Sales': 3, 'It': 2}     "IT" has become "It"!
#
# title() capitalises the first letter of each word and LOWERS the rest, so an
# acronym is mangled. Cleaning can introduce errors as easily as it removes them.

staff["dept"] = staff["dept"].str.strip().str.lower()     # the safe normalisation
print("with .str.lower():", staff["dept"].value_counts().to_dict())
#   -> {'sales': 3, 'it': 2}     consistent, and nothing invented

# The answer to "what is the true average" is: 55,000 from 4 of 5 staff.
# A report that just says 55,000 is hiding a missing person.
#
# Normalising to lower case is the standard move: it is predictable and never
# guesses at capitalisation. The .str methods are Day 1's string methods applied
# to a whole column - Day 18 covers data cleaning properly.
````

### D2. Two libraries, one dataset
For the array below, compute the mean and standard deviation in **both** NumPy and pandas,
and make them agree.

````python
data = np.array([12., 15., np.nan, 18., 21.])

# your code here
````

````python
data = np.array([12., 15., np.nan, 18., 21.])
s = pd.Series(data)

print("--- the defaults disagree ---")
print("numpy  mean:", data.mean(), "| std:", data.std())
print("pandas mean:", s.mean(), "| std:", round(s.std(), 4))

print("\n--- make them agree ---")
# Match pandas to NumPy's behaviour...
print("pandas, numpy-style :", s.mean(skipna=False), "|", s.std(skipna=False))

# ...or match NumPy to pandas' behaviour
print("numpy, pandas-style :", np.nanmean(data), "|", round(np.nanstd(data, ddof=1), 4))
print("pandas default      :", s.mean(), "|", round(s.std(), 4))

# TWO independent differences, both of which must be matched:
#   1. nan handling   numpy propagates   pandas skips     -> nanmean / skipna
#   2. ddof default   numpy 0            pandas 1         -> ddof=1
#
# Fix only one and the numbers still will not agree.
````

## Wrap-up — running the last 5 minutes

Three cold-call questions:

1. *"Same array with a `nan` — NumPy mean and pandas mean?"* → `nan` versus a number.
2. *"`df[["python"]]` — Series or DataFrame?"* → DataFrame. Double brackets always.
3. *"Why not `df.count`?"* → it is the method, not the column.

**Common misconceptions to watch for today**

| Misconception | Correction |
|---|---|
| "pandas and NumPy behave the same" | They differ on `nan` handling and on `ddof` |
| "pandas warns about missing values" | It skips them **silently** |
| "`df["a", "b"]` selects two columns" | That is one tuple key. Use `df[["a", "b"]]` |
| "`df.col` always works" | Not for `count`, `sum`, `mean`, or names with spaces |
| "Looping over rows is fine" | It works, and is hundreds of times slower |
| "Column names are forgiving" | Case-sensitive, and trailing spaces count |
| "`read_csv` keeps integers as integers" | One blank cell makes the column float |
