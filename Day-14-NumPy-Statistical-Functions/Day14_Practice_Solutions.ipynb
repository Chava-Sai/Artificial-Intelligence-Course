# Part A — Predict the Output

### A1. Which is bigger, and by how much?

````python
salaries = np.array([30., 32., 35., 33., 31., 500.])
print(salaries.mean().round(2))
print(np.median(salaries))
````

*Your prediction:*  

> ### ✅ Answer A1
> ```
> 110.17
> 32.5
> ```
> **Why:** the mean uses **every** value, so the single 500 drags it from about 32 up to 110.
> The median only cares about the middle **position**, so it barely moves.
>
> Five of these six people earn about 32. The mean describes **nobody** in the dataset —
> which is the whole argument for reporting the median on skewed data.

### A2. What do these give?

````python
a = np.arange(1., 11.)
print(np.percentile(a, 50))
print(np.median(a))
print(np.ptp(a))
````

*Your prediction:*  

> ### ✅ Answer A2
> ```
> 5.5
> 5.5
> 9.0
> ```
> **Why:** `percentile(a, 50)` **is** the median — they are the same calculation. With ten
> values there is no single middle, so NumPy interpolates between the 5th and 6th (5 and 6)
> to give 5.5.
>
> `ptp` is "peak to peak" — `max - min`, so 10 − 1 = 9. It uses only two values, which means
> one outlier defines it completely.

### A3. What is the IQR?

````python
a = np.arange(1., 11.)
q1, q3 = np.percentile(a, [25, 75])
print(q1, q3)
print(q3 - q1)
````

*Your prediction:*  

> ### ✅ Answer A3
> ```
> 3.25 7.75
> 4.5
> ```
> **Why:** Q1 is the value a quarter of the way through, Q3 three quarters. The IQR is the
> spread of the **middle half** of the data.
>
> Outliers sit outside the middle half by definition, so they cannot distort the IQR the way
> they distort the range or the standard deviation. That is why it is the robust choice —
> and why it defines the whiskers on a box plot.

### A4. What is printed?

````python
d = np.array([1., 2., np.nan, 4.])
print(d.mean())
print(d.sum())
print(d.max())
````

*Your prediction:*  

> ### ✅ Answer A4
> ```
> nan
> nan
> nan
> ```
> **Why:** any arithmetic touching a `nan` produces a `nan`. One missing value out of four
> turns **every** summary into `nan`.
>
> This is deliberate. NumPy refuses to quietly invent an answer — it would rather hand you
> an obviously broken result than a plausible wrong one. Deciding what should happen instead
> is your job, not the library's.

### A5. Which comparison finds the nan?

````python
d = np.array([1., 2., np.nan, 4.])
print(np.nan == np.nan)
print(d == np.nan)
print(np.isnan(d))
````

*Your prediction:*  

> ### ✅ Answer A5
> ```
> False
> [False False False False]
> [False False  True False]
> ```
> **Why:** `nan` is **not equal to itself**. "Unknown" cannot equal "unknown" — you do not
> know whether the two missing values were the same thing.
>
> So `== np.nan` never matches anything, and a filter built on it silently finds nothing.
> **`np.isnan()` is the only correct way to detect a missing value.**

### A6. What do the nan-aware versions give?

````python
d = np.array([1., 2., np.nan, 4.])
print(np.nanmean(d).round(4))
print(np.nanmedian(d))
print(np.nansum(d))
````

*Your prediction:*  

> ### ✅ Answer A6
> ```
> 2.3333
> 2.0
> 7.0
> ```
> **Why:** each function skips the `nan` and works with `[1, 2, 4]`. The mean is 7/3 = 2.333,
> the median of three values is the middle one (2.0), and the sum is 7.
>
> `d[~np.isnan(d)].mean()` gives exactly the same answer — the difference is that filtering
> yourself lets you **report how many** you dropped.

### A7. What happens when everything is missing?

````python
all_nan = np.array([np.nan, np.nan])
print(np.nansum(all_nan))
print(np.nanmean(all_nan))
````

*Your prediction:*  

> ### ✅ Answer A7
> ```
> 0.0
> nan          (plus a RuntimeWarning)
> ```
> **`nansum` returns 0.0, not nan.** A sum of *nothing* is mathematically defined as zero,
> so NumPy reports 0.0 for a column that is entirely missing.
>
> **This is the most dangerous behaviour on today's list.** A real zero and a completely
> missing column look **identical** in your output — and unlike a `nan`, a 0.0 does not
> announce that anything is wrong.
>
> `nanmean` has no such fallback (the mean of nothing is undefined), so it returns `nan` and
> warns.
>
> **The lesson:** always report the count alongside any nan-aware summary.

### A8. What does corrcoef return?

````python
hours  = np.array([1., 2., 3., 4., 5.])
scores = np.array([52., 55., 61., 68., 74.])
print(np.corrcoef(hours, scores).round(4))
print(np.corrcoef(hours, scores)[0, 1].round(4))
````

*Your prediction:*  

> ### ✅ Answer A8
> ```
> [[1.     0.9922]
>  [0.9922 1.    ]]
> 0.9922
> ```
> **Why:** `corrcoef` returns a **matrix**, not a single number. The diagonal is always 1 —
> every variable correlates perfectly with itself. The value you want is the off-diagonal
> element, `[0, 1]`.
>
> `r = 0.99` means hours and scores move together almost perfectly **in a straight line**.
> It does **not** mean studying causes the score, and it would miss a strong curved
> relationship entirely.

### B1. This should count the missing values.

````python
d = np.array([1., 2., np.nan, 4., np.nan])

missing = (d == np.nan).sum()
print("missing values:", missing)
````

*What's wrong:* 

*Your fix:*

> ### ✅ Answer B1
> **Symptom:** prints `0`. No error — it simply reports that nothing is missing, when two
> values are.
> **Cause:** `nan` is not equal to itself, so `== np.nan` matches nothing, ever.
>
> A count of zero looks like good news, which is exactly why this bug survives.

````python
d = np.array([1., 2., np.nan, 4., np.nan])

print("wrong :", (d == np.nan).sum())      # 0
print("right :", np.isnan(d).sum())        # 2

# The full set of checks worth running on any new column:
print()
print("how many :", np.isnan(d).sum())
print("any?     :", np.isnan(d).any())
print("fraction :", np.isnan(d).mean())    # True counts as 1 - Day 1
````

### B2. This should give the class average.

````python
marks = np.array([88., 71., np.nan, 64., 95.])
print("class average:", marks.mean())
````

*What's wrong:* 

*Your fix:*

> ### ✅ Answer B2
> **Symptom:** prints `nan`.
> **Cause:** one missing mark poisons the whole calculation. This one at least **announces**
> itself — a `nan` in your output is obviously wrong, which is kinder than a silently
> incorrect number.

````python
marks = np.array([88., 71., np.nan, 64., 95.])

# Option 1 - the nan-aware function
print("nanmean :", np.nanmean(marks).round(2))

# Option 2 - filter, so you can report what you dropped
clean = marks[~np.isnan(marks)]
print(f"mean of {clean.size} of {marks.size} marks: {clean.mean():.2f}")

# Both give 79.50. The second is the honest one - it makes clear that the
# average is over 4 students, not 5.
````

### B3. This should print the correlation.

````python
hours  = np.array([1., 2., 3., 4., 5.])
scores = np.array([52., 55., 61., 68., 74.])

r = np.corrcoef(hours, scores)
print(f"correlation: {r:.3f}")
````

*Error:* 

*Your fix:*

> ### ✅ Answer B3
> **Error:** `TypeError: unsupported format string passed to numpy.ndarray.__format__`
> **Cause:** `corrcoef` returns a 2×2 **matrix**, and you cannot format a matrix with
> `:.3f`.

````python
hours  = np.array([1., 2., 3., 4., 5.])
scores = np.array([52., 55., 61., 68., 74.])

matrix = np.corrcoef(hours, scores)
print("the full matrix:")
print(matrix.round(4))

r = matrix[0, 1]                      # THE FIX - take the off-diagonal
print(f"\ncorrelation: {r:.3f}")

# The diagonal is always 1 (each variable with itself), and the matrix is
# symmetric, so [0, 1] and [1, 0] are the same value.
````

### B4. This should report the sample standard deviation and the top scorer.

````python
marks = np.array([88., 71., 64., 95., 52.])

print("std        :", marks.std())
print("top scorer :", marks.argmax())
````

*What's wrong:* 

*Your fix:*

> ### ✅ Answer B4
> **Two bugs, neither raising an error.**
> 1. `std()` uses `ddof=0` — the **population** formula. A sample needs `ddof=1`.
> 2. `argmax()` returns the **position** (3), not the mark (95).
>
> Both print plausible numbers, which is what makes them dangerous.

````python
marks = np.array([88., 71., 64., 95., 52.])

print("population std :", marks.std().round(4))
print("sample std     :", marks.std(ddof=1).round(4))
print()
print("position of top:", marks.argmax())
print("the top mark   :", marks.max())
print("both together  :", f"student {marks.argmax()} scored {marks[marks.argmax()]:.0f}")

# Remember NumPy's own inconsistency: std and var default to ddof=0,
# but np.cov defaults to ddof=1. pandas uses ddof=1 for everything.
````

### C2. Fill, and measure the cost
Fill the missing values below with the mean of the values that are present. Then print the
mean and standard deviation **before and after**, and say what changed.

````python
marks = np.array([88., 71., np.nan, 64., 95., np.nan, 52.])

# your code here
````

````python
marks = np.array([88., 71., np.nan, 64., 95., np.nan, 52.])

missing = np.isnan(marks)
fill_value = np.nanmean(marks)

filled = marks.copy()               # .copy() - Day 12! do not modify the original
filled[missing] = fill_value

print("fill value :", round(fill_value, 2))
print()
print(f"before  mean {np.nanmean(marks):.2f}   std {marks[~missing].std(ddof=1):.2f}")
print(f"after   mean {filled.mean():.2f}   std {filled.std(ddof=1):.2f}")

# before  mean 74.00   std 17.54
# after   mean 74.00   std 14.32
#
# WHAT CHANGED: the mean is identical - filling with the mean cannot move it.
# But the STANDARD DEVIATION dropped from 17.54 to 14.32, because the two
# filled values sit exactly at the centre and add no spread at all.
#
# Mean-filling makes data look MORE consistent than it really is. That is the
# cost, and it is invisible unless you check. Day 19 covers this properly.
````

### C3. Find the outlier bounds
Using the 1.5 × IQR rule, print the lower and upper bounds for the data below, then print
any values that fall outside them.

````python
data = np.array([12., 15., 14., 13., 16., 15., 14., 95., 13.])

# your code here
````

````python
data = np.array([12., 15., 14., 13., 16., 15., 14., 95., 13.])

q1, q3 = np.percentile(data, [25, 75])
iqr = q3 - q1
lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

print(f"Q1 {q1}  Q3 {q3}  IQR {iqr}")
print(f"bounds: below {lower} or above {upper}")

outliers = data[(data < lower) | (data > upper)]
print("outliers:", outliers)

# Q1 13.0  Q3 15.0  IQR 2.0
# bounds: below 10.0 or above 18.0
# outliers: [95.]
#
# Note the & / | rule from Day 12 - each condition needs its own brackets.
#
# Compare what the outlier does to each measure:
print()
print(f"mean   {data.mean():.2f}  vs median {np.median(data):.1f}")
print(f"std    {data.std(ddof=1):.2f}  vs IQR   {iqr:.1f}")
# The mean and std are wrecked by one value; the median and IQR are not.
````

For the array below (4 students × 3 subjects, with gaps), print for **each subject**: how

### C5. Correlate, and check it honestly
For the two variables below, print the correlation. Then plot nothing — instead, print the
values and judge whether a single number fairly describes the relationship.

````python
x = np.array([-3., -2., -1., 0., 1., 2., 3.])
y = x ** 2

# your code here
````

````python
x = np.array([-3., -2., -1., 0., 1., 2., 3.])
y = x ** 2

r = np.corrcoef(x, y)[0, 1]
print("r =", round(float(r), 6))
print()
print("x:", x)
print("y:", y)

# r = 0.0
#
# AND YET y IS EXACTLY DETERMINED BY x. Every value of y is x squared -
# a perfect relationship with no noise whatsoever.
#
# Correlation measures only STRAIGHT-LINE relationships. As x rises from -3
# to 0, y falls; as x rises from 0 to 3, y rises. The two halves cancel out
# exactly, giving r = 0.
#
# THE LESSON: r = 0 does not mean "no relationship". It means "no LINEAR
# relationship". Always look at the data before trusting a single number -
# which is why Days 21 and 22 are about plotting.
````

### D1. A reusable describe()
Write `describe(a, name)` that prints a full honest summary of a 1-D array: count, missing,
mean, median, std, range and IQR — and handles an array where **everything** is missing
without crashing.

````python
marks = np.array([88., 71., np.nan, 64., 95., np.nan])

# your code here
````

````python
def describe(a, name="data"):
    """Summarise an array honestly, missing values included."""
    missing = np.isnan(a)
    n_missing = missing.sum()
    clean = a[~missing]

    if clean.size == 0:                       # the guard that matters
        print(f"{name}: no usable values ({a.size} missing)")
        return

    q1, q3 = np.percentile(clean, [25, 75])
    print(f"{name}")
    print(f"  count   {clean.size} of {a.size}   ({n_missing} missing)")
    print(f"  mean    {clean.mean():.2f}")
    print(f"  median  {np.median(clean):.2f}")
    print(f"  std     {clean.std(ddof=1):.2f}")
    print(f"  range   {clean.min():.0f} to {clean.max():.0f}")
    print(f"  IQR     {q3 - q1:.2f}")


describe(np.array([88., 71., np.nan, 64., 95., np.nan]), "marks")
print()
describe(np.array([np.nan, np.nan]), "all missing")
print()
describe(np.array([12., 15., 14., 13., 16., 95.]), "with an outlier")

# The empty-array guard is what stops this crashing on a column that turned
# out to be entirely blank - which happens in real datasets more often than
# you would expect.
#
# Notice in the third call how far the mean sits from the median. That gap
# is the signal to investigate before reporting anything.
````

### D2. Where nansum lies
Build a 2-D array where one column is entirely missing. Compute the column sums with
`np.nansum`, then show why the result is misleading and fix the report.

````python
data = np.array([[1., np.nan, 3.],
                 [4., np.nan, 6.],
                 [7., np.nan, 9.]])

# your code here
````

````python
data = np.array([[1., np.nan, 3.],
                 [4., np.nan, 6.],
                 [7., np.nan, 9.]])

sums = np.nansum(data, axis=0)
print("nansum per column:", sums)

# nansum per column: [12.  0. 18.]
#
# COLUMN 1 REPORTS 0.0 - and a reader has no way to tell that apart from a
# column of genuine zeros. A sum of nothing is defined as zero, so nansum
# has nothing to complain about.

print()
print("compare with a column of REAL zeros:")
zeros = np.array([[1., 0., 3.], [4., 0., 6.], [7., 0., 9.]])
print("nansum       :", np.nansum(zeros, axis=0))
print("  -> identical output, completely different data")

# THE FIX: always report the count alongside the sum
print()
n_present = (~np.isnan(data)).sum(axis=0)
for i, (s, n) in enumerate(zip(sums, n_present)):
    if n == 0:
        print(f"  column {i}: NO DATA")
    else:
        print(f"  column {i}: sum {s:.0f} over {n} values")

# This is the habit worth keeping: a summary statistic without its count
# is not a finding, it is a guess with a decimal point.
````

## Wrap-up — running the last 5 minutes

Three cold-call questions:

1. *"Why does `d == np.nan` find nothing?"* → `nan` is not equal to itself.
2. *"What does `np.nansum` return for an all-missing column?"* → `0.0`, not `nan`.
3. *"`corrcoef` gives you what?"* → a matrix. Take `[0, 1]`.

**Common misconceptions to watch for today**

| Misconception | Correction |
|---|---|
| "The mean is the average" | It is *an* average. One outlier moves it a long way |
| "`== np.nan` finds missing values" | It never matches. Use `np.isnan` |
| "A `nan` will be ignored" | It propagates — every summary becomes `nan` |
| "`nansum` returns `nan` if all are missing" | It returns `0.0`, indistinguishable from real zeros |
| "`r = 0` means no relationship" | It means no **linear** relationship |
| "`r` shows what causes what" | It shows nothing about cause |
| "`std()` is the sample std" | Population. `ddof=1` for sample |

**The habit to enforce today** is reporting the count with every summary. Students find it
fussy until the first time a mean over 4 of 200 rows reaches a slide — after that they do it
without being asked.

**C5 is worth doing on the board** if time allows. A perfect parabola giving `r = 0.0`
surprises everyone, and it is the cleanest possible argument for plotting data before
trusting a statistic. It sets up Days 21 and 22 directly.
