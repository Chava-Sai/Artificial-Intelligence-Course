# 1. Lists

## 1.1 Creating and indexing

A list holds many values **in order**, allows **duplicates**, and can be **changed**.

````python
marks  = [88, 71, 64, 90, 55]
mixed  = [1, "two", 3.0, True]      # any types can share a list
empty  = []
nested = [[1, 2], [3, 4]]           # lists inside lists
chars  = list("PYTHON")             # build one from a string

print(marks)
print(mixed)
print(nested)
print(chars)
print(type(marks), len(marks))
````

### Positions run forwards from 0 and backwards from -1

```
 marks    88    71    64    90    55
 forward   0     1     2     3     4
 negative -5    -4    -3    -2    -1
```

````python
marks = [88, 71, 64, 90, 55]

print("first      :", marks[0])
print("third      :", marks[2])
print("last       :", marks[-1])      # the safe way - never needs len()
print("second-last:", marks[-2])
print("length     :", len(marks))
````

````python
# Going past the end is an error
try:
    print(marks[5])
except IndexError as e:
    print("IndexError:", e)
````

````python
# Changing an item by position
marks = [88, 71, 64, 90, 55]
marks[0] = 100
print(marks)              # [100, 71, 64, 90, 55]

# Nested lists need two indexes
grid = [[1, 2], [3, 4]]
print(grid[1])            # [3, 4]
print(grid[1][0])         # 3
````

## 1.2 Slicing — `[start:stop:step]`

`stop` is **excluded**, exactly like `range()` on Day 2. One rule, not two.

````python
marks = [88, 71, 64, 90, 55]

print("marks[1:4] ->", marks[1:4])    # positions 1,2,3  (not 4)
print("marks[:3]  ->", marks[:3])     # from the start
print("marks[2:]  ->", marks[2:])     # to the end
print("marks[:]   ->", marks[:])      # the whole thing - a COPY
print("marks[-2:] ->", marks[-2:])    # last two
````

````python
# The third number is the step
marks = [88, 71, 64, 90, 55]

print("every 2nd      :", marks[::2])
print("from 1, every 2:", marks[1::2])
print("reversed       :", marks[::-1])      # the neat way to reverse

# Works on strings too, since a string is also a sequence
print("olleh test     :", "hello"[::-1])
````

### Slicing never raises `IndexError`

````python
marks = [88, 71, 64, 90, 55]

print(marks[2:99])     # [64, 90, 55]  - no error
print(marks[99:])      # []            - no error

# Convenient, but it hides mistakes: an empty result may mean your
# indexes were wrong, not that the data was empty.
````

## 1.3 List methods — adding and removing

| Method | What it does |
|---|---|
| `append(x)` | add one item at the end |
| `insert(i, x)` | add at position `i` |
| `extend(other)` | add **all** items of another list |
| `remove(x)` | delete the first matching **value** |
| `pop(i)` | remove by **position** and return it |
| `clear()` | empty the list |

````python
nums = [1, 2, 3]

nums.append(4)          ; print("append(4)    ->", nums)
nums.insert(0, 9)       ; print("insert(0, 9) ->", nums)
nums.extend([7, 8])     ; print("extend([7,8])->", nums)
nums.remove(2)          ; print("remove(2)    ->", nums)   # by VALUE
taken = nums.pop(0)     ; print("pop(0)       ->", nums, "| returned", taken)
````

### ⚠️ `append` vs `extend`

````python
a = [1, 2]
a.append([3, 4])
print("append:", a)      # [1, 2, [3, 4]]  - nested!

b = [1, 2]
b.extend([3, 4])
print("extend:", b)      # [1, 2, 3, 4]    - flat
````

### ⚠️ These all change the list **in place** and return `None`

This is Day 3's "no return means `None`" showing up again.

````python
nums = [1, 2, 3]
result = nums.append(4)

print("the list  :", nums)      # [1, 2, 3, 4]  - it worked
print("the result:", result)    # None          - append returns nothing
````

````python
# So this DESTROYS your list:
nums = [1, 2, 3]
nums = nums.append(4)     # nums is now None
print(nums)

# Correct: just call it, do not reassign
nums = [1, 2, 3]
nums.append(4)
print(nums)
````

## 1.4 Sorting — the clearest example of today's question

````python
# sort() CHANGES the original and returns None
nums = [3, 1, 2]
nums.sort()
print("after sort():", nums)          # [1, 2, 3]

nums = [3, 1, 2]
result = nums.sort()
print("returned    :", result)        # None !
````

````python
# sorted() leaves the original alone and gives you a NEW list
nums = [3, 1, 2]
result = sorted(nums)

print("new list :", result)     # [1, 2, 3]
print("original :", nums)       # [3, 1, 2]  untouched

# sorted() works on ANY iterable, not just lists
print(sorted("hello"))
print(sorted({3, 1, 2}))
print(sorted((3, 1, 2)))
````

````python
# Both take the same two options
nums = [3, 1, 2]
print(sorted(nums, reverse=True))          # largest first

students = [("Ravi", 88), ("Sara", 91), ("Amit", 76)]
print(sorted(students, key=lambda s: s[1]))              # by score
print(sorted(students, key=lambda s: s[1], reverse=True))# best first
````

> `sorted()` is usually the safer default. If you keep the original you can always sort
> again — but an in-place `sort()` has thrown the original order away for good.

## 1.5 Copying a list — and the trap you have met three times

````python
a = [1, 2, 3]
b = a               # a SECOND NAME, not a copy

b.append(4)
print("a:", a)      # [1, 2, 3, 4]  - a changed too!
print("b:", b)
print("same object?", a is b)      # True
````

```
    a ──┐
        ├──►  [1, 2, 3, 4]      one object, two names
    b ──┘
```

### Three ways to make a real copy

````python
a = [1, 2, 3]

b = a[:]          # slice the whole thing
c = a.copy()      # clearest to read
d = list(a)       # rebuild it

b.append(99)
print("a:", a)    # [1, 2, 3]  - untouched
print("b:", b)    # [1, 2, 3, 99]
print("a is b?", a is b)      # False - separate objects
print("a == c?", a == c)      # True  - equal contents
````

> ### This is the fourth time you have met this idea
>
> - **Day 1:** `c = a`, then `c.append()` changed `a`
> - **Day 3:** the mutable default argument that kept growing
> - **Day 3:** a function that appended to the caller's list
> - **Today:** `b = a`
>
> Every one is the same sentence: **two names, one object.**
> It is not four rules to memorise. It is one rule seen from four angles.

````python
# ONE CAVEAT: .copy() is a SHALLOW copy. Nested lists are still shared.
a = [[1, 2], [3, 4]]
b = a.copy()

b[0].append(99)
print("a:", a)      # [[1, 2, 99], [3, 4]]  - the inner list was shared!

# For nested structures, use deepcopy
import copy
a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0].append(99)
print("a:", a)      # [[1, 2], [3, 4]]  - genuinely independent now
````

# 2. Tuples

A tuple is a list that **cannot be changed** after it is created.
Round brackets instead of square.

````python
point  = (3, 7,8,8,)
rgb    = (255, 128, 0)
mixed  = (1, "two", 3.0)

print(point, type(point))
print("point[0]  :", point[0])       # indexing works
print("point[0:1]:", point[0:1])     # slicing works, gives a tuple
print("len       :", len(point))
````

### ⚠️ The single-element tuple needs a trailing comma

````python
single     = (5,)      # a tuple
not_tuple  = (5)       # just the integer 5 in brackets

print(single, type(single))
print(not_tuple, type(not_tuple))
````

### Everything that would change it fails

````python
point = (3, 7)

try:
    point[0] = 9
except TypeError as e:
    print("TypeError:", e)

try:
    point.append(9)
except AttributeError as e:
    print("AttributeError:", e)
````

### Why lock a collection on purpose?

- It **signals intent**: this is not meant to change
- It cannot be corrupted by accident from elsewhere in the program
- It can be a **dictionary key**; a list cannot

### You have used tuples already, without naming them

````python
# Day 1 - the swap
p, q = 5, 9
p, q = q, p
print(p, q)

# Day 3 - returning several values
def stats(nums):
    return min(nums), max(nums)
print(stats([3, 9, 4]), type(stats([3, 9, 4])))

# Day 2 - enumerate gives tuples
for pair in enumerate(["a", "b"]):
    print(pair)

# Day 3 - *args IS a tuple
def f(*args):
    return type(args).__name__
print(f(1, 2, 3))
````

## 2.1 Unpacking

````python
x, y = (3, 7)
print(x, y)

a, b, c = [1, 2, 3]          # works on lists too
print(a, b, c)

first, *rest = [1, 2, 3, 4]  # * collects the remainder
print(first, rest)

try:
    a, *b = [1, 2, 3]
    print(a, b, c)
except ValueError as e:
    print("ValueError:", e)
````

## 2.2 ⚠️ Immutable does not mean frozen inside

````python
t = (1, [2, 3])

try:
    t[1] = [9]           # cannot REPLACE the slot
except TypeError as e:
    print("TypeError:", e)

t[1].append(4)           # but CAN change the list that is in the slot
print(t)                 # (1, [2, 3, 4])
````

A tuple guarantees its slots always point at the **same objects**. It says nothing about
whether those objects can themselves change.

That is why a tuple containing a list cannot be used as a dictionary key — Python cannot
rely on it staying the same. If you need a genuinely fixed collection, keep only
immutable things inside: numbers, strings, and other tuples.

# 3. Sets

A set holds **unique**, **unordered** values.

````python
s = {3, 1, 2, 3, 1}
print(s)                # {1, 2, 3}  - duplicates dropped automatically
print(type(s), len(s))
````

### ⚠️ An empty set is `set()`, not `{}`

````python
print(type(set()))     # <class 'set'>
print(type({}))        # <class 'dict'>  - curly braces alone mean dict!
````

````python
s = {1, 2, 3}

s.add(4)          ; print("add(4)     ->", s)
s.discard(9)      ; print("discard(9) ->", s, "  (no error if missing)")

try:
    s.remove(9)
except KeyError as e:
    print("remove(9) -> KeyError:", e)
````

### ⚠️ No order means no indexing

````python
s = {1, 2, 3}

try:
    print(s[0])
except TypeError as e:
    print("TypeError:", e)

# You CAN loop over it and test membership
for item in s:
    print(item, end=" ")
print()
print("2 in s?", 2 in s)
````

### The everyday use: removing duplicates

````python
names = ["Ravi", "Sara", "Ravi", "Amit", "Sara"]

lost_order = list(set(names))              # fast, but order is LOST
kept_order = list(dict.fromkeys(names))    # order preserved (Python 3.7+)

print("original :", names)
print("set()    :", lost_order)
print("fromkeys :", kept_order)
````

> `list(set(x))` is the quick way to deduplicate, but it silently throws away the original
> order. With rows of data, order almost always matters — use `dict.fromkeys()` instead.

### Membership is dramatically faster on a set

````python
import time

big_list = list(range(1_000_000))
big_set  = set(big_list)

t = time.time(); 999_999 in big_list; list_time = time.time() - t
t = time.time(); 999_999 in big_set;  set_time  = time.time() - t

print(f"list: {list_time*1000:.3f} ms")
print(f"set : {set_time*1000:.3f} ms")
print("A list scans every item. A set jumps straight to the answer.")
````

## 3.1 Set algebra

With `a = {1, 2, 3}` and `b = {3, 4, 5}`:

| Operator | Method | Meaning | Result |
|---|---|---|---|
| `a \| b` | `a.union(b)` | everything in either | `{1,2,3,4,5}` |
| `a & b` | `a.intersection(b)` | only what is in both | `{3}` |
| `a - b` | `a.difference(b)` | in a but not in b | `{1,2}` |
| `a ^ b` | `a.symmetric_difference(b)` | in one but not both | `{1,2,4,5}` |

````python
a = {1, 2, 3}
b = {3, 4, 5}

print("union        a | b :", a | b)
print("intersection a & b :", a & b)
print("difference   a - b :", a - b)
print("difference   b - a :", b - a)      # note: not the same!
print("symmetric    a ^ b :", a ^ b)
````

````python
# A realistic framing
section_a = {"ravi", "sara", "amit"}
section_b = {"sara", "priya"}

print("in both sections :", section_a & section_b)
print("all attendees    :", section_a | section_b)
print("only in A        :", section_a - section_b)
print("in exactly one   :", section_a ^ section_b)
````

# 4. Choosing between them

| | list | tuple | set |
|---|---|---|---|
| Written as | `[1, 2, 3]` | `(1, 2, 3)` | `{1, 2, 3}` |
| Keeps order | yes | yes | no |
| Can be changed | yes | no | yes (add / remove) |
| Allows duplicates | yes | yes | no |
| Index and slice | yes | yes | no |
| Can be a dict key | no | yes* | no |
| Membership test | slow on big data | slow on big data | fast, any size |

\* only if everything inside it is immutable too

**Default to a list.** Reach for a tuple when the data is a fixed record, and a set when
you need uniqueness or fast membership.

````python
# Conversions
print(list((1, 2, 3)))
print(tuple([1, 2, 3]))
print(set([1, 2, 2, 3]))
print(list({3, 1, 2}))       # order not guaranteed
print(list("abc"))
````

### Three patterns worth memorising

````python
# 1. Remove duplicates, keep order
names = ["Ravi", "Sara", "Ravi", "Amit"]
print(list(dict.fromkeys(names)))
````

````python
# 2. Fast membership filter - build the set ONCE, outside the loop
valid_ids = set([101, 102, 103])          # built once

rows = [(101, "Ravi"), (999, "Ghost"), (103, "Amit")]
for rid, name in rows:
    if rid in valid_ids:                   # fast, however big the set
        print("keep:", name)
````

# 5. Putting it together — cleaning a messy attendance list

````python
raw      = ["Ravi", "sara", "RAVI", "Amit", "Sara", "", "Priya", "Kiran"]
enrolled = {"ravi", "sara", "amit", "priya", "neha"}


def clean(names):
    """Normalise names and drop blanks, keeping order."""
    out = []
    for n in names:
        n = n.strip().lower()       # tidy the text
        if n:                       # skip empty strings
            out.append(n)
    return out


present = clean(raw)
unique  = list(dict.fromkeys(present))    # dedupe, keep order

absent  = enrolled - set(unique)          # on the roll but not here
unknown = set(unique) - enrolled          # here but not on the roll

print("Present :", unique)
print("Absent  :", sorted(absent))
print("Unknown :", sorted(unknown))
````

What each piece contributes:

- **List** — `raw` and `present` keep order
- **Set** — `enrolled` gives uniqueness and fast lookup
- **`dict.fromkeys`** — deduplicates *without* losing order
- **Set difference** — finds who is missing in one line
- **`sorted()`** — returns a new list, leaving the roll untouched
- **Day 3's function** — one job, a docstring, and it returns a value

Doing the `absent` calculation with lists alone would need a nested loop. The set does it
in one line.

# 6. Recap — the twelve things to remember

1. `list [ ]` ordered and changeable · `tuple ( )` locked · `set { }` unique.
2. Index from `0` forwards, from `-1` backwards.
3. Slices exclude the stop value, exactly like `range()`.
4. `[::-1]` reverses; `[:]` makes a copy.
5. `append` adds one item; `extend` adds all of them.
6. In-place methods return `None` — never reassign their result.
7. `sort()` changes the list; `sorted()` returns a new one.
8. `b = a` is a second name. `b = a.copy()` is a copy.
9. A single-item tuple needs the trailing comma: `(5,)`
10. A tuple is locked, but a list inside it is not.
11. `set()` for an empty set — `{}` is an empty dict.
12. Sets: no order, no index, no duplicates, fast lookup.
