### A1. What does this build?

````python
nums = [1, 2, 3, 4]print([n * 2 for n in nums])
````

*Your prediction:*  

> ### ✅ Answer A1> ```> [2, 4, 6, 8]> ```> **Why:** the expression `n * 2` runs once per item, and every result goes into the new list.> No condition, so nothing is dropped — four in, four out.>> The equivalent loop is three lines: create an empty list, loop, append.

### A2. What is the result — and its type?

````python
words = ["python", "is", "great"]print([len(w) for w in words])print({len(w) for w in words})
````

*Your prediction:*  

> ### ✅ Answer A2> ```> [6, 2, 5]> {2, 5, 6}> ```> **Why:** square brackets build a **list** — three items, in order.> Curly brackets build a **set** — unordered, and duplicates would be dropped.>> Here the lengths happen to be distinct, so both have three items. Ask what> `{len(w) for w in ["ab", "cd"]}` would give → `{2}`, a single item.

### A3. How many items come out?

````python
nums = [3, -1, 4, -5, 2]print([n for n in nums if n > 0])
````

*Your prediction:*  

> ### ✅ Answer A3> ```> [3, 4, 2]> ```> **Three items from five.** The `if` at the **end** is a **filter** — the two negative> numbers are simply not included.>> Read it as: *"keep n, for each n, IF it is positive."*

### A4. Same data. How many items now?

````python
nums = [3, -1, 4, -5, 2]print(["pos" if n > 0 else "neg" for n in nums])
````

*Your prediction:*  

> ### ✅ Answer A4> ```> ['pos', 'neg', 'pos', 'neg', 'pos']> ```> **Five items from five.** The `if`/`else` at the **front** is a **choice**, not a filter —> every item produces something, so nothing is dropped.>> **A3 vs A4 is the whole lesson:**> | Position | Effect | Length |> |---|---|---|> | `if` at the end | drops items | may be shorter |> | `if`/`else` at the front | changes items | always the same |>> Ask the class which they would use to *"replace negatives with zero"* → front> (`[n if n > 0 else 0 for n in nums]`), because nothing should be dropped.

### A5. What does this produce?

````python
d = {"a": 1, "b": 2, "c": 3}print({k: v * 10 for k, v in d.items() if v > 1})
````

*Your prediction:*  

> ### ✅ Answer A5> ```> {'b': 20, 'c': 30}> ```> **Why:** three things happening at once —> - `.items()` gives `(key, value)` pairs, unpacked into `k` and `v`> - `if v > 1` at the end **filters**, so `"a"` is dropped> - `k: v * 10` builds each surviving entry with the value scaled>> The colon is what makes this a dict rather than a set.

### A6. What is `flat`?

````python
matrix = [[1, 2], [3, 4], [5, 6]]flat = [n for row in matrix for n in row]print(flat)
````

*Your prediction:*  

> ### ✅ Answer A6> ```> [1, 2, 3, 4, 5, 6]> ```> **Why:** two `for` clauses in **one** set of brackets give a **flat** list. They read in> the same order as the nested loops you would otherwise write — outer first, inner second.>> Contrast with `[[n for n in row] for row in matrix]`, which uses a whole comprehension> as the expression and gives back `[[1, 2], [3, 4], [5, 6]]` — a nested result.

### A7. What prints — both lines?

````python
g = (n for n in [1, 2, 3])print(sum(g))print(sum(g))
````

*Your prediction:*  

> ### ✅ Answer A7> ```> 6> 0> ```> **Why:** round brackets make a **generator**, not a tuple and not a list. A generator> hands out its values once and is then **exhausted** — the second `sum()` finds nothing> left and returns 0.>> There is no error. Just a wrong answer, silently. That is what makes it worth knowing.>> A list can be reused as many times as you like: `lst = [1,2,3]; sum(lst); sum(lst)` → `6`, `6`.

### A8. What is `result`, and what is printed?

````python
nums = [1, 2, 3]result = [print(n) for n in nums]print("result:", result)
````

*Your prediction:*  

> ### ✅ Answer A8> ```> 1> 2> 3> result: [None, None, None]> ```> **Why:** `print()` displays its argument and **returns `None`** (Day 3's lesson). So the> comprehension dutifully collects three `None` values into a list nobody wanted.>> **The rule:** if you are not building a collection, write a `for` loop. Using a> comprehension purely for a side effect wastes memory and misleads the reader.

---# Part B — Spot & Fix the Bug  *(7 min)*

---# Part B — Spot & Fix the Bug**Teaching note:** B1 and B2 are the same confusion approached from opposite sides.Do them as a pair.

### B1. This should keep only the positives.

````python
nums = [3, -1, 4, -5]print([n if n > 0 for n in nums])
````

*Error:* *Your fix:*

> ### ✅ Answer B1> **Error:** `SyntaxError: expected 'else' after 'if' expression`> **Cause:** a bare `if` at the **front** is not allowed. At the front you are writing a> conditional expression, which must always say what to build in *both* cases.> For filtering, the `if` belongs at the **end**.

````python
nums = [3, -1, 4, -5]print([n for n in nums if n > 0])         # filter - if at the END# If you actually wanted to REPLACE negatives rather than drop them,# the front position is right - but then you must supply an else:print([n if n > 0 else 0 for n in nums])  # [3, 0, 4, 0]
````

### B2. This should replace negatives with zero.

````python
nums = [3, -1, 4, -5]print([n for n in nums if n > 0 else 0])
````

*Error:* *Your fix:*

> ### ✅ Answer B2> **Error:** `SyntaxError`> **Cause:** the opposite mistake to B1. There is no `else` in the filtering position —> a filter either keeps an item or does not. To *choose* between two values you must move> the whole `if`/`else` to the **front**.

````python
nums = [3, -1, 4, -5]print([n if n > 0 else 0 for n in nums])     # [3, 0, 4, 0]  - same length# B1 and B2 are one idea from two sides:#   if at the END      -> filter, no else possible#   if/else at FRONT   -> choose, else REQUIRED
````

### B3. This should give a tuple of squares.

````python
nums = [1, 2, 3]squares = (n * n for n in nums)print(squares)print(squares[0])
````

*Error:* *Your fix:*

> ### ✅ Answer B3> **Symptom:** the first print shows `<generator object ...>`, then> `TypeError: 'generator' object is not subscriptable`.> **Cause:** round brackets do **not** make a tuple comprehension — there is no such thing.> They make a generator, which has no positions to index.

````python
nums = [1, 2, 3]squares = tuple(n * n for n in nums)     # convert the generator to a tupleprint(squares, type(squares).__name__)print(squares[0])# Or build a list, which is what you usually want anyway:squares = [n * n for n in nums]print(squares, squares[0])
````

### B4. This should print each name. Something extra happens.

````python
names = ["Ravi", "Sara"]output = [print(n) for n in names]print(output)
````

*What's wrong:* *Your fix:*

> ### ✅ Answer B4> **Symptom:** the names print, but then `[None, None]` appears.> **Cause:** a comprehension always **builds a list**. Since `print()` returns `None`,> you get a list of `None` values that serves no purpose.> On a large dataset this quietly allocates memory for nothing.

````python
names = ["Ravi", "Sara"]for n in names:          # no list is built - this is the right tool    print(n)# The rule: comprehensions are for BUILDING a collection.# If you only want a side effect, use a for loop.
````

### C2. Filter and transform at onceFrom the list below, produce a list of the **uppercase** names that arelonger than three characters.

````python
names = ["ravi", "jo", "sara", "al", "priya"]# your code here
````

### C3. Grade every markTurn the marks below into a list of grades — `"pass"` for 40 or more, `"fail"` otherwise.The result must have the **same number of items** as the input.

````python
marks = [88, 35, 71, 22, 64]# your code here
````

````python
marks = [88, 35, 71, 22, 64]grades = ["pass" if m >= 40 else "fail" for m in marks]print(grades)          # ['pass', 'fail', 'pass', 'fail', 'pass']print(len(marks), len(grades))    # 5 5 - nothing dropped# The "same number of items" requirement is the clue that this is a CHOICE,# not a filter - so the if/else goes at the FRONT.
````

### C4. Invert a dictionarySwap the keys and values of the dictionary below, in one line.

````python
codes = {"in": "India", "us": "USA", "uk": "Britain"}# your code here
````

````python
codes = {"in": "India", "us": "USA", "uk": "Britain"}inverted = {v: k for k, v in codes.items()}print(inverted)        # {'India': 'in', 'USA': 'us', 'Britain': 'uk'}# Yesterday this was a five-line loop.# Same caveat as yesterday: if two keys shared a value, one would be lost.
````

### C5. Clean a messy listFrom the raw list below, produce a clean list of names: stripped of spaces,title-cased, with the blanks removed.

````python
raw = ["  ravi ", "SARA", "", "  ", "amit  "]# your code here
````

````python
raw = ["  ravi ", "SARA", "", "  ", "amit  "]clean = [n.strip().title() for n in raw if n.strip()]print(clean)           # ['Ravi', 'Sara', 'Amit']# Note .strip() appears twice - once to test, once to transform.# That is the one small inefficiency of this very common idiom, and it is# almost always worth it for the readability. On a huge dataset you would# use a loop and strip once into a variable.
````

### D2. Transpose a matrixTurn rows into columns:`[[1, 2, 3], [4, 5, 6]]`  →  `[[1, 4], [2, 5], [3, 6]]`Do it with a nested comprehension. (Then check your answer against `zip`.)

````python
matrix = [[1, 2, 3], [4, 5, 6]]# your code here
````

````python
matrix = [[1, 2, 3], [4, 5, 6]]# The loop version first - this is what the comprehension compressestransposed = []for j in range(len(matrix[0])):        # for each COLUMN position    new_row = []    for row in matrix:                 # take that position from every row        new_row.append(row[j])    transposed.append(new_row)print(transposed)# The nested comprehension - same two loops, same ordertransposed = [[row[j] for row in matrix] for j in range(len(matrix[0]))]print(transposed)      # [[1, 4], [2, 5], [3, 6]]# The inner comprehension builds one column; the outer repeats it per position.# Nested brackets -> a nested result (contrast with A6's flattening).# And the way you would actually write it in real code:print([list(col) for col in zip(*matrix)])# zip(*matrix) pairs up the matching positions from every row - Day 4's zip,# with Day 3's * unpacking the rows into separate arguments.
````
