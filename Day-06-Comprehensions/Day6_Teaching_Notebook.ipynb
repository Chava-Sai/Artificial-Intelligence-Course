---# 1. List comprehensions## 1.1 The transformationHere is the loop you already write:```pythonsquares = []for n in numbers:    squares.append(n * n)```And the same thing as one line:```pythonsquares = [n * n for n in numbers]#           ▲       ▲      ▲#           │       │      └── WHAT to loop over#           │       └───────── the loop variable#           └───────────────── WHAT to build (from the append call)```Read it as: **"build n\*n, for each n in numbers."**The pieces are identical — only the order changed.

````python
numbers = [1, 2, 3, 4]# the loopsquares = []for n in numbers:    squares.append(n * n)print(squares)# the comprehensionsquares = [n * n for n in numbers]print(squares)
````

## 1.2 The basic form```[  expression   for   item   in   iterable  ]```The square brackets say *"build me a list"*. Everything inside says what to put in it.

````python
nums  = [1, 2, 3]names = ["ravi", "sara"]words = ["python", "is", "great"]raw   = ["88", "71", "64"]print([n * n for n in nums])              # square every numberprint([n.upper() for n in names])         # uppercase every nameprint([len(w) for w in words])            # length of each wordprint([int(x) for x in raw])              # cast a column to intprint([n for n in nums])                  # just a copy - Day 4's list(a)
````

## 1.3 Adding a condition — keeping only some items```[  expression   for   item   in   iterable   if   condition  ]```The `if` goes at the **end**. Items that fail the test are simply not included.

````python
marks = [88, 35, 71, 22, 64]# the looppassed = []for m in marks:    if m >= 40:        passed.append(m)print(passed)# the comprehension - four lines to onepassed = [m for m in marks if m >= 40]print(passed)
````

````python
nums  = [1, 2, 3, 4, 5, 6]words = ["hi", "python", "is", "great"]raw   = [" Ravi ", "", "sara ", "  "]print([n for n in nums if n % 2 == 0])          # even numbers onlyprint([w for w in words if len(w) > 3])         # longer than three lettersprint([n.strip() for n in raw if n.strip()])    # clean AND drop the blanks
````

That last one does both jobs at once: it **transforms** with `.strip()` and **filters**out anything that is empty after stripping.

## 1.4 ⚠️ `if` / `else` — the one that trips everybodyThe position of the `if` changes what it means.

````python
nums = [3, -1, 4, -5]# FILTERING - if at the END. Some items are DROPPED.print([n for n in nums if n > 0])                    # [3, 4]  - shorter!# CHOOSING - if/else at the FRONT. Nothing is dropped.print(["pos" if n > 0 else "neg" for n in nums])     # 4 items - same length
````

| Position | What it does | Length of the result ||---|---|---|| `if` at the **end** | **drops** items | may be shorter || `if`/`else` at the **front** | **changes** items | always the same |**The way to remember it:** ask whether you want a *shorter* list or a *different* list.The position follows from the answer.

````python
# You CANNOT put a bare if at the front - uncomment to see the error# [n if n > 0 for n in nums]#   -> SyntaxError# At the front you are writing Day 1's conditional expression, which ALWAYS needs# an else: Python must know what to build for every single item.# At the end you are filtering, and no else is possible.print("Uncomment the line above to see the SyntaxError.")
````

````python
# You can combine both - filter AND choosemarks = [88, 35, 71, 22, 64]print(["A" if m >= 75 else "B" for m in marks if m >= 40])#       ^^^^^^^^^^^^^^^^^^^^^^         ^^^^^^^^^^^^^^^^#       choose (front)                 filter (end)# Only the passes are included, and each one gets a grade.
````

---# 2. Dict and set comprehensions## 2.1 Dictionary comprehensions```{  key_expression : value_expression   for   item   in   iterable  }```Curly brackets **and a colon**. The colon is what makes it a dict rather than a set.

````python
marks = {"ravi": 88, "sara": 35, "amit": 64}# Yesterday's five-line filter loop...passed = {}for k, v in marks.items():    if v >= 40:        passed[k] = vprint(passed)# ...becomes one line, exactly as promisedpassed = {k: v for k, v in marks.items() if v >= 40}print(passed)
````

````python
d     = {"a": 1, "b": 2}words = ["python", "is", "great"]print({k: v * 2 for k, v in d.items()})       # scale the valuesprint({v: k for k, v in d.items()})           # INVERT itprint({w: len(w) for w in words})             # word -> lengthprint({n: n * n for n in range(1, 6)})        # number -> squareprint(dict(zip(["a", "b"], [1, 2])))          # two lists -> one dict
````

> The inversion line was a five-line loop in yesterday's practice. It carries the same> warning: if two keys share a value, one wins and the other is silently lost.

## 2.2 Set comprehensionsCurly brackets, **no colon**.

````python
print({n * n for n in [1, 2, 2, 3]})          # {1, 4, 9} - duplicates droppedprint({w.lower() for w in ["Ravi", "RAVI", "Sara"]})print({len(w) for w in ["python", "is", "great"]})   # the distinct lengths
````

### ⚠️ There is no tuple comprehension

````python
g = (n * n for n in [1, 2, 3])print(g)                    # a GENERATOR, not a tupleprint(type(g).__name__)# If you want a tuple, build a list and convert itprint(tuple([n * n for n in [1, 2, 3]]))
````

### The four brackets, side by side| You write | You get | Because ||---|---|---|| `[n for n in x]` | a **list** | square brackets || `{n for n in x}` | a **set** | curly brackets, no colon || `{k: v for k, v in x}` | a **dict** | curly brackets **with** a colon || `(n for n in x)` | a **generator** | round brackets |

---# 3. Going further## 3.1 Two loops inside one comprehension

````python
matrix = [[1, 2], [3, 4], [5, 6]]# the loopflat = []for row in matrix:    for n in row:        flat.append(n)print(flat)# the comprehension - the two for clauses appear in the SAME orderflat = [n for row in matrix for n in row]print(flat)
````

````python
# ⚠️ Reversing the two for clauses fails - the inner variable does not exist yettry:    flat = [n for n in row for row in matrix]except NameError as e:    print("NameError:", e)
````

## 3.2 A comprehension inside a comprehension

````python
# Two for clauses in ONE bracket -> a FLAT listflat = [i * j for i in range(1, 4) for j in range(1, 4)]print("flat:", flat)# A whole comprehension as the expression -> a NESTED listtable = [[i * j for j in range(1, 4)] for i in range(1, 4)]print("grid:", table)for row in table:    print(row)
````

The distinction matters: **two `for` clauses give you a flat list**, while**nested brackets give you a list of lists**.

## 3.3 Generator expressions

````python
# A list comprehension builds everything immediatelysquares_list = [n * n for n in range(10)]print(squares_list)# A generator expression builds nothing until askedsquares_gen = (n * n for n in range(10))print(squares_gen)print(list(squares_gen))      # now it produces the values
````

````python
# The common use: feeding straight into a function.# The brackets are optional when it is the only argument.print(sum(n * n for n in range(1, 5)))print(max(len(w) for w in ["python", "is", "great"]))print(any(n < 0 for n in [1, 2, -3]))
````

### ⚠️ A generator is used up after one pass

````python
g = (n for n in [1, 2, 3])print(sum(g))     # 6print(sum(g))     # 0 !  it is exhausted - and there is no error to warn you# A list can be reused as many times as you likelst = [1, 2, 3]print(sum(lst), sum(lst))
````

Use a list unless you have a specific reason not to. Generators matter when the data islarge enough that holding it all in memory is a problem — today this isrecognition-level knowledge.

---# 5. Putting it together — a cleaning script, rewritten

````python
raw   = [" Ravi ", "sara", "", "AMIT", " priya"]marks = [88, 71, 0, 64, 92]# ---- BEFORE: 20 lines ----names = []for n in raw:    n = n.strip().title()    if n:        names.append(n)book = {}for i in range(len(names)):    book[names[i]] = marks[i]passed = {}for k, v in book.items():    if v >= 40:        passed[k] = vgrades = []for v in passed.values():    if v >= 75:        grades.append("A")    else:        grades.append("B")print(names)print(book)print(passed)print(grades)
````

````python
# ---- AFTER: 4 lines ----names2  = [n.strip().title() for n in raw if n.strip()]book2   = dict(zip(names2, marks))passed2 = {k: v for k, v in book2.items() if v >= 40}grades2 = ["A" if v >= 75 else "B" for v in passed2.values()]print(names2)print(book2)print(passed2)print(grades2)# Identical resultsprint("\nall match:", names == names2 and book == book2      and passed == passed2 and grades == grades2)
````

What each line uses:- **Filter + transform** — strip, title-case, and drop the blanks in one pass- **`zip`** — two lists into one dictionary- **`if` at the end** — keep only the passes- **`if`/`else` at the front** — one grade per student, nothing droppedSame result, same logic — but a reader now takes in each step at a glance instead oftracing an accumulator.
