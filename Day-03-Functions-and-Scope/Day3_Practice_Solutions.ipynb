### A4. Run three times. What does each call print?

````python
def add_item(item, basket=[]):    basket.append(item)    return basketprint(add_item("apple"))print(add_item("bread"))print(add_item("milk"))
````

### A6. Three lines print. Which value is on each?

````python
x = "global"def outer():    x = "enclosing"    def inner():        x = "local"        print(x)    inner()    print(x)outer()print(x)
````

*Your prediction:*  

> ### ✅ Answer A6> ```> local> enclosing> global> ```> **Why — the LEGB rule.** Python looks a name up as **L**ocal → **E**nclosing → **G**lobal> → **B**uilt-in, and stops at the first match.>> Each `x` shadows the one outside it, but none of them overwrite each other — the three> assignments create three separate names in three separate scopes.>> This is also why naming a variable `list` or `sum` breaks things: your Global name is> found before Python's Built-in.

### B4. Every call should start with an empty list. It doesn't.

````python
def collect(value, results=[]):    results.append(value)    return resultsprint(collect(1))print(collect(2))print(collect(3))
````

*What's wrong:* *Your fix:*

> ### ✅ Answer B4> **Symptom:** `[1]`, then `[1, 2]`, then `[1, 2, 3]` — the list keeps growing.> **Cause:** the mutable default trap. The default list is created once at definition time> and shared by every call that omits the argument.

````python
def collect(value, results=None):    if results is None:        results = []          # a fresh list on every call    results.append(value)    return resultsprint(collect(1))     # [1]print(collect(2))     # [2]print(collect(3))     # [3]# Passing your own list still works, and is still yours to keep:mine = []collect(10, mine)collect(20, mine)print("mine:", mine)          # [10, 20]
````

### D1. Factorial — two waysWrite `factorial(n)` returning `n!` (so `factorial(5)` is `120`).Do it **once with a loop**, then again **calling itself** (recursion).Remember `0! = 1`.

````python
# your code here
````

````python
# WAY 1 - a loop (the accumulator pattern from Day 2)def factorial_loop(n):    """Return n! using a loop."""    result = 1    for i in range(2, n + 1):        result *= i    return result# WAY 2 - recursion: the function calls itself on a smaller problemdef factorial(n):    """Return n! using recursion."""    if n <= 1:            # BASE CASE - stops the recursion        return 1    return n * factorial(n - 1)     # RECURSIVE CASEfor n in [0, 1, 5, 10]:    print(f"{n}! = {factorial_loop(n)}  /  {factorial(n)}")# How factorial(5) unwinds:#   5 * factorial(4)#     4 * factorial(3)#       3 * factorial(2)#         2 * factorial(1)  -> base case returns 1#   = 5 * 4 * 3 * 2 * 1 = 120## Every recursive function needs a base case. Without one you get# RecursionError: maximum recursion depth exceeded.
````

Flexible calculatorWrite `calculate(*numbers, operation="sum")` that accepts any count of numbers and returns:- the total when `operation="sum"`- the product when `operation="product"`- the average when `operation="average"`- the string `"unknown operation"` for anything elseTest all four cases.

````python
# your code here
````

````python
def calculate(*numbers, operation="sum"):    """Combine any number of values according to the chosen operation.    Args:        *numbers:  the values to combine        operation: "sum", "product" or "average" (default "sum")    Returns:        The computed result, or "unknown operation" if the name is not recognised.    """    if not numbers:        return 0                       # guard against an empty call    if operation == "sum":        return sum(numbers)    elif operation == "product":        result = 1        for n in numbers:            result *= n        return result    elif operation == "average":        return sum(numbers) / len(numbers)    return "unknown operation"print(calculate(1, 2, 3, 4))                          # 10print(calculate(1, 2, 3, 4, operation="product"))     # 24print(calculate(1, 2, 3, 4, operation="average"))     # 2.5print(calculate(1, 2, 3, 4, operation="median"))      # unknown operationprint(calculate())                                    # 0# Note: `operation` MUST be passed as a keyword. Anything positional after *numbers# gets swallowed by *numbers, so calculate(1, 2, "product") would try to multiply a string.
````
