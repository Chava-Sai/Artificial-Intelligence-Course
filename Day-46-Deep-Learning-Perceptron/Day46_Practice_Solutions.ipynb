Run top to bottom in Colab. These answers follow the practice notebook question order.

```python
import numpy as np
import matplotlib.pyplot as plt

def step(z):
    """Our Day 46 convention: 1 at or above zero, else 0."""
    return int(z >= 0)

def perceptron(x, w, b):
    z = float(np.dot(x, w) + b)
    return z, step(z)

w = np.array([0.8, 0.6])
b = -1.0
print('Ready. weights:', w, 'bias:', b)
```

## Question 2 · Work by hand

Contributions are 0.7 and -0.4. z = 0.7 - 0.4 + 0.2 = 0.5. Output = 1.

```python
print(perceptron([1,1], [0.7,-0.4], 0.2))
```

## Question 3 · Other rows

[0,0] → z=0.2, output 1; [0,1] → z=-0.2, output 0; [1,0] → z=0.9, output 1. A positive bias makes even the all-zero row activate.

```python
for x in [[0,0],[0,1],[1,0]]: print(x, perceptron(x,[0.7,-0.4],0.2))
```

## Question 4 · Write the function

Calculate the dot product, add bias, and apply the step. Return both values so we can inspect the decision.

```python
def predict(x,w,b):
    z = float(np.dot(x,w)+b)
    return z, int(z >= 0)
assert predict([1,1],[0.7,-0.4],0.2) == (0.49999999999999994,1) or np.isclose(predict([1,1],[0.7,-0.4],0.2)[0],.5)
print(predict([1,1],[0.7,-0.4],0.2))
```

## Question 5 · Move the bias

```python
print(perceptron([1,1],[0.7,-0.4],-0.4))
```

## Question 6 · Boundary
