### A5. What happens to the mean and the spread?

```python
s = pd.Series([10., 20., np.nan, 40., 50.])
print(round(s.mean(), 2), round(s.std(), 2), s.count())
f = s.fillna(s.mean())
print(round(f.mean(), 2), round(f.std(), 2), f.count())
```
