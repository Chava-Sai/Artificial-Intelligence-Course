### A1. How many numbers print, and which?

````python
for i in range(1, 5):    print(i)
````

*Your prediction:*  

> ### ✅ Answer A1> ```> 1> 2> 3> 4> ```> **Why:** `range(1, 5)` stops **before** 5. Four numbers, not five.> To get 1 through 5 you need `range(1, 6)`.> This is the off-by-one bug, and it is the most common loop error there is.

### A2. What is the final value of `total`?

````python
total = 0for n in [5, 10, 15]:    total += nprint(total)
````

*Your prediction:*  

> ### ✅ Answer A2> ```> 30> ```> **Trace:** `0` → `+5` = `5` → `+10` = `15` → `+15` = `30`>> This is the accumulator pattern: initialise before the loop, update inside it, use it after.> Ask the class where `print(total)` would need to be to print `5, 15, 30` instead — one indent deeper.

### A3. Which numbers print?

````python
for n in [1, 2, 3, 4, 5]:    if n == 3:        continue    if n == 5:        break    print(n)
````

*Your prediction:*  

> ### ✅ Answer A3> ```> 1> 2> 4> ```> **Why:**> - `1`, `2` → print normally> - `3` → `continue` skips the print, loop **carries on**> - `4` → prints> - `5` → `break` exits before reaching the print>> The distinction to hammer: `continue` skips **one item**, `break` stops the **whole loop**.

### A4. What does this print — and how many lines?

````python
for i in range(1, 4):    for j in range(1, 3):        print(i, j)
````

*Your prediction:*  

> ### ✅ Answer A4> ```> 1 1> 1 2> 2 1> 2 2> 3 1> 3 2> ```> **Six lines** — 3 outer passes × 2 inner passes.>> The inner loop runs **completely** on every outer pass, and `j` restarts from 1 each time.> Nested loop cost **multiplies**: this is why 1000 × 1000 is a million iterations.

### A5. What grade prints? Look carefully at the keywords.

### B2. ⚠️ This runs forever. Predict why, then fix it.*(Run it if you like — then press the stop button. That is the lesson.)*

````python
count = 1while count <= 5:    print(count)    # something is missing here
````

*What's wrong:* *Your fix:*

> ### ✅ Answer B2> **Symptom:** the cell never finishes — it prints `1` endlessly.> **Cause:** `count` is never updated, so `count <= 5` is always `True`.> **The three parts of a while loop:** initialise, condition, **update**. The update was missing.

````python
count = 1while count <= 5:    print(count, end=" ")    count += 1          # THE FIX - without this the condition never becomes Falseprint()print("Finished, count is", count)
````

### B3. Every student gets grade "C". Why?

````python
m = 88grade = Noneif m >= 75:    grade = "A"if m >= 60:    grade = "B"if m >= 40:    grade = "C"print(grade)
````

*What's wrong:* *Your fix:*

> ### ✅ Answer B3> **Symptom:** wrong grade, **no error message**.> **Cause:** three separate `if`s. All are true for 88, so each overwrites the previous one.> **Fix:** `elif` — Python then stops at the first match.

````python
m = 88if m >= 75:    grade = "A"elif m >= 60:      # only tested if the first was False    grade = "B"elif m >= 40:    grade = "C"else:    grade = "F"print(grade)       # A# Also check the ORDER: putting >= 40 first would give "C" for a 90.
````

### B4. This should print a 3x3 grid. It prints one number per line.

````python
for i in range(1, 4):    for j in range(1, 4):        print(f"{i*j:4}", end="")        print()
````

*What's wrong:* *Your fix:*

> ### ✅ Answer B4> **Symptom:** nine lines with one number each, instead of a 3×3 grid.> **Cause:** the bare `print()` is indented **inside** the inner loop, so it starts a new line> after every single number. It belongs in the **outer** loop, where it ends each row.> **Indentation is the logic here, not decoration.**

````python
for i in range(1, 4):    for j in range(1, 4):        print(f"{i*j:4}", end="")    print()          # moved out one level - runs once per ROW
````

````python
# your code here
````
