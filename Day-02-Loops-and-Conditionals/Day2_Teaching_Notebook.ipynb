---# 1. Conditionals## 1.1 Indentation defines the blockOther languages use `{ }`. Python uses **indentation**. The colon `:` opens a block;the indent tells Python what is inside it.```if condition:          <- colon opens the block    do_this()          <- 4 spaces = inside    and_this()         <- also insidedo_this_always()       <- not indented = outside, always runs```

````python
marks = 55if marks >= 40:    print("Pass")           # inside the if    print("Well done")      # also insideprint("Result printed")     # OUTSIDE - runs no matter what
````

````python
# Same code, one line moved. Completely different meaning.marks = 20if marks >= 40:    print("Pass")print("Well done")          # now OUTSIDE - prints even when failing!
````

> ### The rules> - 4 spaces is the standard. Colab inserts them automatically after a colon.> - Be **consistent**. Never mix tabs and spaces — that gives a `TabError`.> - Wrong indentation is either an `IndentationError` (loud, easy) or **silently wrong logic** (quiet, dangerous).

````python
# Uncomment to see the error# if marks >= 40:# print("Pass")        # IndentationError: expected an indented blockprint("Uncomment the lines above to see the error.")
````

## 1.3 `elif` — picking exactly one from manyPython checks conditions **top to bottom** and stops at the **first** one that is `True`.

````python
percentage = 82if percentage >= 75:    grade = "A"elif percentage >= 60:    grade = "B"elif percentage >= 40:    grade = "C"else:    grade = "F"print(f"{percentage}% -> grade {grade}")
````

### ⚠️ Order matters — and getting it wrong produces no error at all

````python
percentage = 90# WRONG ORDER: least restrictive firstif percentage >= 40:    grade = "C"          # 90 >= 40 is True, so Python stops HEREelif percentage >= 60:    grade = "B"elif percentage >= 75:    grade = "A"print("Wrong order gives:", grade)      # "C" for a 90! No error, just wrong.
````

### ⚠️ Separate `if`s are NOT an `elif` chain

````python
m = 90# BUG: every condition is tested, so later ones overwrite earlier onesgrade = Noneif m >= 75: grade = "A"if m >= 60: grade = "B"       # also True -> overwrites "A"if m >= 40: grade = "C"       # also True -> overwrites "B"print("Separate ifs give:", grade)     # "C"# CORRECT: elif stops after the first matchif m >= 75: grade = "A"elif m >= 60: grade = "B"elif m >= 40: grade = "C"print("elif chain gives: ", grade)     # "A"
````

## 2.3 `enumerate` and `zip` — the two that make loops readable

````python
names = ["Ravi", "Sara", "Amit"]# Clumsy: index-basedfor i in range(len(names)):    print(i, names[i])print("---")# Idiomatic: enumerate gives you both at oncefor i, name in enumerate(names):    print(i, name)print("---")# Count from 1 instead of 0for i, name in enumerate(names, start=1):    print(f"{i}. {name}")
````

## 2.4 The accumulator patternThe single most important loop shape in data work. Three steps:1. **Initialise** a variable before the loop2. **Loop** over the data3. **Update** the variable inside the loop

````python
marks = [45, 78, 33, 90, 62]total = 0                 # 1. INITIALISEfor m in marks:           # 2. LOOP    total = total + m     # 3. UPDATE   (or: total += m)print("Total:", total)print("Average:", total / len(marks))
````

---# 3. `while` loopsA `for` loop runs a known number of times. A `while` loop runs **until a condition stops being true**.Three parts you must always write:1. **Initialise** the variable the condition depends on2. Write the **condition**3. **Update** that variable inside the loop — miss this and the loop never ends

````python
count = 1                  # 1. INITIALISEwhile count <= 5:          # 2. CONDITION    print(count)    count += 1             # 3. UPDATE   <- without this it runs foreverprint("Done, count is now", count)
````

## 3.1 `for` or `while`? Ask one question> **"Do I know, before I start, how many times this should run?"**>> Yes → `for`  ·  No → `while`| Use `for` | Use `while` ||---|---|| Every row in a dataset | Until the user types valid input || Every character in a string | Until a game ends || Exactly 10 times | Until a value converges (model training!) || Every file in a folder | Until a server responds |In practice you will write `for` loops perhaps nine times out of ten.

## 3.2 The classic `while`: refusing bad input

````python
marks = -1        # start invalid so the loop runs at least oncewhile marks < 0 or marks > 100:    raw = input("Enter marks (0-100): ")    if raw.isdigit():        marks = int(raw)        if marks > 100:            print("Must be 100 or less.")    else:        print("Digits only, please.")print(f"Recorded: {marks}")
````

**Why this shape works:**- `marks` starts at `-1`, so the condition is `True` on the first check — the loop always runs once- The condition describes what is **wrong**, so the loop continues *while* the input is bad- `.isdigit()` screens the text before `int()` sees it, so no `ValueError`- It exits only when a genuinely valid number has been storedThis is Day 1's `input()` lesson applied: screen the string, then cast it.

---# 4. `break`, `continue`, `pass`## 4.1 `break` — leave the loop entirely, right now

````python
for n in [1, 2, 3, 4, 5]:    if n == 3:        break    print(n)          # prints 1, 2 then stops completelyprint("Loop exited")
````

````python
# Real use: stop as soon as you find what you were looking fornames = ["Ravi", "Sara", "Amit", "Priya"]target = "Amit"for i, name in enumerate(names):    if name == target:        print(f"Found {target} at position {i}")        break            # no point checking the rest
````

## 4.2 `continue` — skip this item, carry on with the loop

````python
for n in [1, 2, 3, 4, 5]:    if n == 3:        continue    print(n)          # prints 1, 2, 4, 5 - the loop does NOT stopprint("Loop finished normally")
````

````python
# Real use: filter out items you do not care aboutmarks = [45, -1, 78, 200, 62]for m in marks:    if m < 0 or m > 100:        continue              # skip invalid entries    print("Valid mark:", m)
````

> ### `break` vs `continue` — the distinction to hold on to> - `break` = **stop the loop**> - `continue` = **skip this one item**, the loop keeps going>> In nested loops, both affect only the **innermost** loop they sit in.

## 4.4 `for` / `else` — the one nobody expectsThe `else` on a loop runs **only if the loop finished without hitting `break`**.Read it as *"for ... if no break"*. It has nothing to do with `if`/`else`.

````python
numbers = [4, 7, 2, 9]for n in numbers:    if n < 0:        print("Found a negative:", n)        breakelse:    print("All values were positive")     # runs - no break happened
````

````python
numbers = [4, -7, 2, 9]for n in numbers:    if n < 0:        print("Found a negative:", n)        breakelse:    print("All values were positive")     # does NOT run - we broke out
````
