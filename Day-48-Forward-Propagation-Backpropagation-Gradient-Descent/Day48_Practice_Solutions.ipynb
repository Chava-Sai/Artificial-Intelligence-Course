```python
import numpy as np
np.set_printoptions(precision=6, suppress=True)
def sigmoid(z): return 1/(1+np.exp(-np.asarray(z,dtype=float)))
```

## 01. First hidden score

$(0.2)(1)+(-0.3)(2)+0.1=-0.3$.

```python
z1_first = .2*1-.3*2+.1
print(z1_first)
```

## 02. Sigmoid output

$\sigma(-0.3)=0.4256$ and $\sigma(0.4)=0.5987$.

```python
a1 = sigmoid(np.array([-.3,.4]))
print(a1.round(4))
```

## 03. Output probability and BCE

The logit is about $0.09855$; $p\approx0.52462$ and BCE is about $0.64509$.

```python
z2 = float(np.array([.7,-.5]) @ sigmoid(np.array([-.3,.4])) + .1)
p = float(sigmoid(z2))
loss = float(np.logaddexp(0,z2)-z2)
print(z2,p,loss)
```

## 04. A scalar chain rule

$\frac{\partial L}{\partial w}=(a-y)a(1-a)x$. Here $a=0.5$, so $(0.5-1)(0.5)(0.5)(2)=-0.25$.

```python
a = float(sigmoid(0.0))
dL_dw = (a-1)*a*(1-a)*2
print(dL_dw)
```

## 11. ReLU derivative

Outputs $[0,0,3]$. Derivatives $[0,0,1]$ using the common convention $g\prime(0)=0$.

```python
z=np.array([-2.,0.,3.])
relu=np.maximum(0,z)
relu_prime=(z>0).astype(float)
print(relu,relu_prime)
```

## 12. Gradient check

Analytic $2(w-3)=-4$ at $w=1$. The centered difference gives approximately $-4$.

```python
f=lambda w:(w-3)**2
eps=1e-5
numeric=(f(1+eps)-f(1-eps))/(2*eps)
analytic=2*(1-3)
print(numeric,analytic)
assert abs(numeric-analytic)<1e-8
```
